# Linear Combinations

If $x_1, x_2, \dots, x_n$ are objects that support scalar multiplication and addition, and $\alpha_1, \alpha_2, \dots, \alpha_n$ are scalars, then

$$
\alpha_1 x_1 + \alpha_2 x_2 + \cdots + \alpha_n x_n
$$

is a linear combination.

These objects can be:
- vectors
- matrices
- polynomials
- functions
- tensors

This generality is one reason linear algebra is so useful in data science and deep learning.

At a high level, a linear combination means:
- take some vectors, matrices, or other mathematical objects
- multiply each one by a scalar
- add the results together

## Why it matters
In machine learning, a prediction is often built as a weighted sum of input features. That weighted sum is a linear combination.
For example, in a simple model:

$$
y = w_1 x_1 + w_2 x_2 + w_3 x_3 + b
$$

This is a linear combination of the input features $x_1, x_2, x_3$, where the weights $w_1, w_2, w_3$   tell the model how important each feature is.

This idea appears in:
- linear regression
- logistic regression
- support vector machines
- principal component analysis
- matrix factorization
- neural networks

In deep learning, every neuron first computes a weighted sum of its inputs before applying an activation function.

$$
z = w_1 x_1 + w_2 x_2 + \cdots + w_n x_n + b
$$
That weighted sum is again a linear combination. So whenever you train a neural network, the model is learning the best coefficients for these linear combinations.

Understanding linear combinations helps you:
- understand how models combine features
- see how neurons work internally
- understand embeddings as combinations of learned directions
- understand image, audio, and text representations
- build intuition for span, basis, subspaces, and vector spaces
- better understand dimensionality reduction methods like PCA
- reason about matrix multiplication and transformations


### Linear regression

$$
y = w_1x_1 + w_2x_2 + \cdots + w_nx_n + b
$$
The prediction is a linear combination of input features.

### Logistic regression

Before applying the sigmoid, the model computes:

$$
z = w_1x_1 + w_2x_2 + \cdots + w_nx_n + b
$$

Again, a linear combination.

### Neural networks

Each neuron computes:

$$
z = Wx + b
$$

This is a linear combination of the input vector components.

### Embeddings

Word embeddings, item embeddings, and user embeddings are often combined using weighted sums or matrix multiplication, which are built from linear combinations.

### Attention in transformers

Attention outputs are weighted combinations of value vectors.
Those weighted sums are linear combinations.

So even in advanced deep learning architectures, this idea remains central.


## What is a linear combination?

If $x_1$ and $x_2$ are two objects, and $\alpha_1$ and $\alpha_2$ are scalars, then $\alpha_1 x_1 + \alpha_2 x_2$ is called a linear combination of $x_1$ and $x_2$.

The scalars $\alpha_1$ and $\alpha_2$ are the coefficients or weights. By changing these coefficients, we can create many different results from the same original objects.


### Example

Let

$$

x_1 =
\begin{bmatrix}
3 \\
5 \\
9
\end{bmatrix},
\quad
x_2 =
\begin{bmatrix}
1 \\
2 \\
1
\end{bmatrix}
$$

We can create different linear combinations by choosing different scalar values.

Example 1

$$
y_1 = 2x_1 + 3x_2
$$

$$
= 2
\begin{bmatrix}
3 \\
5 \\
9
\end{bmatrix}
+
3
\begin{bmatrix}
1 \\
2 \\
1
\end{bmatrix}
=
\begin{bmatrix}
6 \\
10 \\
18
\end{bmatrix}
+
\begin{bmatrix}
3 \\
6 \\
3
\end{bmatrix}
=
\begin{bmatrix}
9 \\
16 \\
21
\end{bmatrix}

$$

Example 2

$$
y_2 = 0x_1 + 0x_2
=
\begin{bmatrix}
0 \\
0 \\
0
\end{bmatrix}
$$

This gives the zero vector.

Example 3

$$
y_3 = -1x_1 + 2x_2

=
-1
\begin{bmatrix}
3 \\
5 \\
9
\end{bmatrix}
+
2
\begin{bmatrix}
1 \\
2 \\
1
\end{bmatrix}
=
\begin{bmatrix}
-3 \\
-5 \\
-9
\end{bmatrix}
+
\begin{bmatrix}
2 \\
4 \\
2
\end{bmatrix}
=
\begin{bmatrix}
-1 \\
-1 \\
-7
\end{bmatrix}

$$


We can generate infinitely many linear combinations from just two vectors if the coefficients come from the real numbers.

Why? Because there are infinitely many possible values for the scalars.

For example:
 - $1.2x_1 + 0.7x_2$
 - $-5x_1 + 10x_2$
 - $0.001x_1 - 8.6x_2$

Each different choice of coefficients creates a new combination. This is important in ML because model parameters are continuous values, so a model can learn many possible combinations of features.


## Geometric intuition in 2D

Suppose $x_1$ and $x_2$ are vectors in 2D space. Then expressions like

$$
x_1 + x_2
\quad \text{or} \quad
-2x_1 - x_2
$$

are also vectors in the same space. Geometrically, a linear combination lets us move in the directions defined by the original vectors.
This is a very important idea because in machine learning, features often define directions in a space, and the model learns how much to move along each direction.

## Linear combinations in RGB images
An RGB image can be understood using linear combinations. Each pixel is represented by three values:

(r, g, b)

where:
 - r = red intensity
 - g = green intensity
 - b = blue intensity

Usually each value is between 0 and 255.

Now define three basic color vectors:

$$
r =
\begin{bmatrix}
1 \\
0 \\
0
\end{bmatrix},
\quad
g =
\begin{bmatrix}
0 \\
1 \\
0
\end{bmatrix},
\quad
b =
\begin{bmatrix}
0 \\
0 \\
1
\end{bmatrix}
$$

Then any pixel color can be written as:

$$
\alpha_1 r + \alpha_2 g + \alpha_3 b
$$

where $\alpha_1, \alpha_2, \alpha_3$ are the red, green, and blue intensity values respectively.

This means every RGB color is a linear combination of the red, green, and blue basis vectors. Why this matters in ML

Images in computer vision are stored as arrays and tensors.
Understanding pixels as linear combinations helps you understand:
- image representation
- convolution operations
- channel mixing
- color transformations
- feature extraction in CNNs




```python
import torch
import matplotlib.pyplot as plt

I, C, n = torch.ones((100, 100)), torch.zeros((100, 100, 3)) , 5

plt.figure(figsize=(20, 10))
for i in range(n):
    rgb = torch.randint(low=0, high=255, size=(3,))
    r, g, b = rgb[0] , rgb[1], rgb[2]
    C[:, :, 0] , C[:, :, 1] , C[:, :, 2] = b*I , g*I, r*I
    plt.subplot(1, n, i+1);plt.imshow(C.byte().numpy());plt.axis('off');
```

## Linear combinations of matrices

Matrices can also be expressed as linear combinations. For example, a $2 \times 2$ matrix

$$
\begin{bmatrix}
a & b \\
c & d
\end{bmatrix}
$$

can be written as:

$$
a
\begin{bmatrix}
1 & 0 \\
0 & 0
\end{bmatrix}
+
b
\begin{bmatrix}
0 & 1 \\
0 & 0
\end{bmatrix}
+
c
\begin{bmatrix}
0 & 0 \\
1 & 0
\end{bmatrix}
+
d
\begin{bmatrix}
0 & 0 \\
0 & 1
\end{bmatrix}

$$

For example:

$$

\begin{bmatrix}
-2 & 4 \\
1 & 9
\end{bmatrix}
=
-2
\begin{bmatrix}
1 & 0 \\
0 & 0
\end{bmatrix}
+
4
\begin{bmatrix}
0 & 1 \\
0 & 0
\end{bmatrix}
+
1
\begin{bmatrix}
0 & 0 \\
1 & 0
\end{bmatrix}
+
9
\begin{bmatrix}
0 & 0 \\
0 & 1
\end{bmatrix}

$$

Why this matters in ML and DL, Matrices are everywhere:
- datasets
- weight matrices in neural networks
- covariance matrices
- attention score matrices
- embedding tables
- transformations between feature spaces

Understanding matrices as combinations of simpler building blocks helps you understand how models are structured.


## Linear combinations in the xy-plane

Any point in 2D space can be written as a linear combination of:

$$
\begin{bmatrix}
1 \\
0
\end{bmatrix}
\quad \text{and} \quad
\begin{bmatrix}
0 \\
1
\end{bmatrix}

$$

For example,

$$

\begin{bmatrix}
4 \\
7
\end{bmatrix}
=
4
\begin{bmatrix}
1 \\
0
\end{bmatrix}
+
7
\begin{bmatrix}
0 \\
1
\end{bmatrix}
$$

This shows that the standard basis vectors can generate every point in the plane. This idea is the starting point for understanding:
- basis
- span
- feature spaces
- embeddings
- latent representations


**A linear combination is just:**

“take some building blocks, scale them, and add them together.”

That is exactly what many ML models do with features, vectors, embeddings, and matrices.
