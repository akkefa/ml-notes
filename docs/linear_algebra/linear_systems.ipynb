# Linear System
A linear system is a collection of linear equations that share the same unknown variables. The goal is to find the values of the variables that satisfy all the equations simultaneously. In machine learning, these unknown variables are often the model parameters or weights we want to learn from data.


Suppose each training example is represented by a feature vector

$$
x_i =
\begin{bmatrix}
x_{i1} \\
x_{i2} \\
\vdots \\
x_{id}
\end{bmatrix}
$$

and has an associated target value $y_i$. If we assume a linear relationship between the input features and the target, then for each data point we write:

$$
w_1x_{i1} + w_2x_{i2} + \cdots + w_dx_{id} = y_i
$$

This is a linear equation in the unknown model parameters $w_1, w_2, \dots, w_d$. When we collect many training examples, we obtain many such equations, which together form a system of linear equations.

This is important because machine learning is often about finding parameter values that explain data. In other words, training a model can be viewed as solving a system that comes from the dataset.

$$
w_1x_{i1} + w_2x_{i2} + \cdots + w_dx_{id} = y_i
$$

This is a linear equation in the unknown model parameters $w_1, w_2, \dots, w_d$. When we collect many training examples, we obtain many such equations, which together form a system of linear equations.

This is important because machine learning is often about finding parameter values that explain data. In other words, training a model can be viewed as solving a system that comes from the dataset.

If the system has an exact solution, then there exists a set of weights that perfectly matches all training examples. However, real-world data is usually noisy, incomplete, or inconsistent, so an exact solution often does not exist. In that case, machine learning finds the best approximate solution by minimizing error.

This is why linear systems are so important: they teach us how to represent data mathematically, how to think about model parameters, and why optimization is needed when exact fitting is impossible.

The concept also helps us understand whether a problem has:

- no solution, meaning the data constraints conflict
- one unique solution, meaning the parameters are fully determined
- infinitely many solutions, meaning the data is not sufficient to determine a single model

These ideas are highly relevant in both machine learning and deep learning. In machine learning, they appear directly in linear regression, least squares, and related methods. In deep learning, each neuron computes a weighted sum of inputs, which is itself a linear equation, and each layer applies a matrix transformation to data before passing it through a nonlinear activation.


## Example

$$
2x + 3y = 5 \\
4x - y = 1
$$

To solve this system, we can use various methods such as substitution, elimination, or matrix operations.

### Substitution Method
 Solve one of the equations for one variable in terms of the other. For example, from the first equation, we can express `y` in terms of `x`:

$$y = \frac{5 - 2x}{3}$$

2. Substitute this expression for `y` into the second equation:

$$4x - \frac{5 - 2x}{3} = 1$$

3. Solve for `x`:

$$4x - \frac{5}{3} + \frac{2x}{3} = 1$$
4. Combine like terms and solve for `x`:

$$\frac{12x}{3} + \frac{2x}{3} = 1 + \frac{5}{3}$$

$$\frac{14x}{3} = \frac{8}{3}$$

$$x = \frac{8}{14}$$

$$x = \frac{4}{7}$$

5. Substitute the value of `x` back into the expression for `y` to find `y`:

$$y = \frac{5 - 2\left(\frac{4}{7}\right)}{3}$$
$$y = \frac{5 - \frac{8}{7}}{3}$$
$$y = \frac{\frac{35}{7} - \frac{8}{7}}{3}$$
$$y = \frac{\frac{27}{7}}{3}$$
$$y = \frac{27}{21}$$
$$y = \frac{9}{7}$$


Thus, the solution to the system of linear equations is `x = 4/7` and `y = 9/7`.

### Elimination Method
 Multiply the equations to align the coefficients of one variable. For example, we can multiply the first equation by 4 and the second equation by 2 to align the coefficients of `x`:

$$
8x + 12y = 20 \\
8x - 2y = 2
$$

2. Subtract the second equation from the first to eliminate `x`:

$$
(8x + 12y) - (8x - 2y) = 20 - 2
$$

$$
14y = 18
$$

$$
y = \frac{18}{14}
$$

$$
y = \frac{9}{7}
$$

3. Substitute the value of `y` back into one of the original equations to find `x`:

$$
2x + 3\left(\frac{9}{7}\right) = 5
$$

$$
2x + \frac{27}{7} = 5
$$

$$
2x = 5 - \frac{27}{7}
$$

$$
2x = \frac{35}{7} - \frac{27}{7}
$$

$$
2x = \frac{8}{7}
$$

$$
x = \frac{8}{7} \div 2
$$

$$
x = \frac{8}{14}
$$

$$
x = \frac{4}{7}
$$

Thus, the solution to the system of linear equations is `x = 4/7` and `y = 9/7`.

### Matrix Method
Write the system of equations in matrix form `Ax = b`, where `A` is the coefficient matrix, `x` is the variable vector, and `b` is the constant vector:

$$A = \begin{bmatrix}2 & 3\\4 & -1\end{bmatrix}, \quad
x = \begin{bmatrix}x\\y\end{bmatrix}, \quad
b = \begin{bmatrix}5\\1\end{bmatrix}$$

2. Use matrix operations to solve for `x`. We can find the inverse of matrix `A` and multiply it by `b` to find `x`:

$$A^{-1} = \frac{1}{\det(A)} \text{adj}(A)$$
Where `\det(A)` is the determinant of `A` and `\text{adj}(A)` is the adjugate of `A`. Calculating the determinant:

$$\det(A) = (2)(-1) - (4)(3) = -2 - 12 = -14$$
Calculating the adjugate:

$$\text{adj}(A) = \begin{bmatrix}-1 & -3\\-4 & 2\end{bmatrix}$$
Calculating the inverse:

$$A^{-1} = \frac{1}{-14} \begin{bmatrix}-1 & -3\\-4 & 2\end{bmatrix}$$
$$A^{-1} = \begin{bmatrix}1/14 & 3/14\\4/14 & -2/14\end{bmatrix}$$
Now, we can find `x = A^{-1} * b`:

$$x = \begin{bmatrix}1/14 & 3/14\\4/14 & -2/14\end{bmatrix} \begin{bmatrix}5\\1\end{bmatrix}$$
Calculating the product:

$$x = \begin{bmatrix}(1/14)(5) + (3/14)(1)\\(4/14)(5) + (-2/14)(1)\end{bmatrix}$$
$$x = \begin{bmatrix}8/14\\18/14\end{bmatrix}$$
$$x = \begin{bmatrix}4/7\\9/7\end{bmatrix}$$

Thus, the solution to the system of linear equations is `x = 4/7` and `y = 9/7`.  

In conclusion, there are multiple methods to solve a system of linear equations, and the choice of method can depend on the specific problem and the preferences of the solver. Each method ultimately leads to the same solution, which in this case is `x = 4/7` and `y = 9/7`.






## Why linear systems matter in ML

Suppose you have data points and want to learn weights $w_1, w_2, \dots, w_d$.

In many ML models, each training example gives a condition like:

$w_1 x_1 + w_2 x_2 + \cdots + w_d x_d = y$

That is a linear equation in the unknown parameters. If you have many training examples, you get many such equations together. That becomes a system of linear equations.

So the topic matters because it teaches:

- how model parameters are related to data
- when exact fitting is possible
- when no exact solution exists
- when many solutions exist
- why optimization is needed
- why matrices are used to represent datasets and weights

In machine learning:

-  the unknowns are the model parameters
-  the equations come from the training examples
-  the solution is the parameter vector that best explains the data


Before you study advanced algorithms, systems of linear equations help you understand:

1. What a model is really doing

A model is searching for weights that satisfy relationships in data.

2. Why datasets are written as matrices

A dataset becomes a matrix X, weights become a vector w, and outputs become a vector y.

So the whole learning problem becomes:

$$
Xw = y
$$


That one equation is one of the most important equations in machine learning.

3. Why exact solutions are rare

Real data is noisy, inconsistent, incomplete, and high-dimensional.
So often there is no exact solution to $Xw=y$.
That is why we minimize loss instead.

4. Why optimization exists

When the system cannot be solved exactly, we try to find the best approximate solution.

That leads to least squares, gradient descent, backpropagation, regularization, and many ML ideas.

## Where systems of linear equations appear in ML

### Linear regression

This is the most direct example. Suppose each data point has features $x_i$ and target $y_i$.
A linear model assumes:

$$
w_1x_{i1} + w_2x_{i2} + \cdots + w_dx_{id} = y_i $$

for each training example i.

If we collect all examples, we get:

$$
Xw = y
$$

where:

- X = data matrix
- w = unknown weights
- y = target vector

Why linear systems matter here Because training linear regression is fundamentally about solving this system. But what happens in practice?

Usually:

- there are more equations than unknowns
- data is noisy
- equations may conflict

So exact equality often does not hold. Then we solve:

$$
\min_w \|Xw - y\|^2$$

This is least squares. And the normal equation becomes:

$$
X^TXw = X^Ty
$$

That is again a system of linear equations.

So even when the original problem is approximate, the solution still relies on linear systems.

### Logistic regression

Logistic regression is not linear in output probability, but inside it still uses a linear score:

$$
z = Xw + b
$$

The model then applies sigmoid:

$$
\hat{y} = \sigma(z)
$$

So even though the final prediction is nonlinear, the core combination of features is linear. Why linear systems matter here

Because they help you understand:

- weighted sums of features
- separating hyperplanes
- matrix-vector multiplication
- parameter estimation structure

Logistic regression training uses optimization, not direct linear solving, but the underlying structure comes from linear equations.

### Neural networks / deep learning

Deep learning looks nonlinear because of activations like ReLU, sigmoid, GELU, softmax. But every layer begins with a linear transformation:

$$
z = Wx + b
$$

or for a batch:

$$
Z = XW + b
$$

This is linear algebra. Each neuron computes a weighted sum of inputs. That weighted sum is just a linear equation.

Why linear systems matter here

Because they teach the foundation of:

- neurons
- fully connected layers
- matrix multiplication
- affine transformations
- parameter learning
- feature mixing

So even deep learning is built on repeated linear operations plus nonlinear activations. A common way to say it is:

Deep learning = many stacked linear transformations + nonlinearities Without understanding linear equations and systems, the “linear transformation” part remains unclear.


### Solving for parameters in closed-form models

Some ML models can be solved exactly or approximately using matrix equations.

Examples:

- linear regression
- ridge regression
- least squares problems
- low-rank approximation
- some recommendation methods
- Kalman filtering
- certain probabilistic models

For example, ridge regression solves:

$$
(X^TX + \lambda I)w = X^Ty
$$

That is a system of linear equations. So this topic directly appears in actual algorithms.

### Geometry of solutions

Systems of linear equations teach you about geometry:

* one equation in 2D = line
* two equations in 2D = intersection of lines
* one equation in 3D = plane
* many equations = intersection of subspaces

This is very important in ML because model fitting is geometric.

Why?

Because learning often means finding where constraints meet.

Examples:

- decision boundaries
- hyperplanes in classification
- feasible solution spaces
- null spaces
- column spaces
- projection onto subspaces

A lot of ML intuition becomes easier once you think geometrically.


### Understanding no solution / one solution / infinite solutions

This part is very useful in machine learning.

No solution: If the system is inconsistent, no exact parameter vector fits all points.

In ML this happens all the time because data is noisy or contradictory.

That is why we use loss minimization instead of exact solving.

One solution: If the data constraints determine a unique parameter vector, the model has an exact unique fit.

Infinite solutions: If the data does not constrain the parameters enough, there are many solutions. This also happens in ML, especially when:

- there are more parameters than constraints
- features are redundant
- columns are linearly dependent
- deep networks are overparameterized

This is extremely important in deep learning, where many parameter settings may fit training data well.


## Why this matters specifically in deep learning

Let’s make this more direct. Deep learning uses millions or billions of parameters.
You may think: “If deep learning is nonlinear, why should I care about linear systems?”

Because:

1. Every layer uses linear algebra,Each layer computes:

$$
Wx + b
$$

That is a structured linear operation.

2. Backprop depends on matrix operations
Gradients are propagated through chains of matrix multiplications and vector operations.

3. Local approximations are often linear
Even nonlinear networks are often analyzed locally using linear approximations.

4. Optimization relies heavily on linear algebra
Gradients, Hessian approximations, Jacobians, projections, curvature approximations — all use linear algebra ideas.

5. Representation learning happens in vector spaces
Embeddings, activations, attention outputs, hidden states — all live in vector spaces shaped by linear transformations.


### Concrete detailed example: linear regression as a linear system

Suppose we want to predict house price from two features:

- $x_1$: house size
- $x_2$: number of rooms

Assume model:

$y = w_1x_1 + w_2x_2$

Now suppose we have 3 data points:

1. size = 1, rooms = 1, price = 3
2. size = 2, rooms = 1, price = 5
3. size = 3, rooms = 1, price = 7

This gives:

$w_1 + w_2 = 3$

$2w_1 + w_2 = 5$

$3w_1 + w_2 = 7$

This is a system of linear equations. Subtract first equation from second:

$w_1 = 2$

Then from first:

$w_2 = 1$

So the learned model is:

$y = 2x_1 + x_2$

Why this example matters It shows that learning parameters can literally be solving a system.

Example with no exact solution

Now suppose the third point is noisy:

1. $w_1 + w_2 = 3$
2. $2w_1 + w_2 = 5$
3. $3w_1 + w_2 = 8$

From first two equations we still get:

$w_1 = 2,\quad w_2 = 1$

But then third equation becomes:

$3(2) + 1 = 7 \neq 8$

So there is no exact solution.

Why this matters in ML

Because real data is noisy. This is why ML usually does not solve exact systems.
Instead it finds the best approximate solution.

That leads directly to loss functions.

For example, minimize:

$$(w_1+w_2-3)^2 + (2w_1+w_2-5)^2 + (3w_1+w_2-8)^2$$

That is machine learning in action.

⸻

Example with infinite solutions

Suppose we have:

$w_1 + w_2 = 4$

This is only one equation with two unknowns.

There are infinitely many solutions:

* $(4,0)$
* $(3,1)$
* $(2,2)$
* $(1,3)$

Why this matters in ML

Because if the dataset does not provide enough independent information, many models can fit equally well.

This happens when:

- too few samples
- redundant features
- parameters > constraints
- overparameterized neural networks

So the concept of infinite solutions is not just math theory. It explains real ML behavior.

Matrix form: the most important ML form

Instead of writing many equations one by one, we write them compactly:

$$
Xw = y
$$

Example:

$$
\begin{bmatrix}
1 & 1 \\
2 & 1 \\
3 & 1
\end{bmatrix}
\begin{bmatrix}
w_1 \\
w_2
\end{bmatrix}
=
\begin{bmatrix}
3 \\
5 \\
7
\end{bmatrix}
$$

This is how datasets are handled in machine learning. Why this form is powerful:

- compact
- scalable
- easy to compute
- works with matrix libraries
- fits naturally with GPUs and vectorized operations

Almost every ML framework uses this kind of batched matrix form.
