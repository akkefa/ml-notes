A function is a relation from a set of inputs to a set of possible outputs, where each input is related to exactly one output.

$$y = f(x)$$

This means that if an object, $x$, is in the set of inputs (called the domain), then a function, $f$, maps the object, $x$, to precisely one object, $f(x)$ in the set of possible outputs (called the codomain).


$$f(x) = x^2$$

We can then use the function $f$ to compute the square of any number. For example, $f(2) = 2^2 = 4$ and $f(3) = 3^2 = 9$.

$$
\mathrm{f}(x) = \sqrt{x + {6}} \\
\mathrm{f}(6) = \sqrt{10 + {6}} \\
\mathrm{f}(6) = 4.0
$$

Example: Array input, scalar output#

$$f([x_1, x_2, \ldots, x_n]) = \sum_{i=1}^{n} x_i$$


````python
import torch

def f(x):
  return torch.sum(x)

print(f(torch.tensor([-1, 3, 4.6, 0, 9])))
````

Example: Array input, array output#
$$f([x_1, x_2, \ldots, x_n]) = [x_1^2, x_2^2, \ldots, x_n^2]$$

````python
def f(x):
  return torch.sort(x)

print(f(torch.tensor([-1, 3, 4.6, 0, 9])))
````

### Visualization of a function
Visualizing a function can be beneficial. In data analysis, visualization can give a better insight into the relationship between input and output.

Visualizing polynomials#
Let’s visualize some simple polynomial functions: 
$$f(x) = x^2$$
$$f(x) = x^3$$


````python
import seaborn as sns
import matplotlib.pyplot as plt
x = torch.arange(0, 5., 0.2)
y1 = x ** 2
y2 = x ** 3

sns.lineplot(x=x, y=y1, label='f(x) = x^2')
sns.lineplot(x=x, y=y2, label='f(x) = x^3')
plt.title('Visualization of Polynomial Functions')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.show()
````

### Functions in Data Science

Understand how data science functions map numerical inputs to outputs using arrays, matrices, and tensors. Learn to approximate unknown functions from sample data and see how linear algebra helps simplify complex relationships through linear function approximation.

1. A face recognition system takes the image of a face as input and outputs a number describing the identity of the face in the image. This function maps a matrix to a number.
2. Speech recognition takes an audio signal as input and outputs a transcript of the spoken words. This function maps a matrix (or tensor) to a sequence of characters or words.
3. Object detection takes an image as input and outputs the locations and classes of objects in the image. This function maps a matrix to a set of bounding boxes and class labels.

### Function approximation in data science
As described in the examples above, the output of a real-world system is an unknown function (say $f(x)$ ) of its input. Our goal in data science is to estimate this function reliably. We don’t know anything about the function except for a few sample inputs and corresponding outputs called training data. Let’s also assume the type of underlying function (say, a a second degree polynomial) based on some data analysis. We still have to estimate its parameters to best relate the inputs to the corresponding outputs (because there may be several second degree polynomials, depending on the parameters).

More formally, considering the given inputs,
​
$$\{x_1, x_2, \ldots, x_n\}$$
and the corresponding outputs,
$$\{y_1, y_2, \ldots, y_n\}$$

$$ \forall i \in \{1,\dots,n\},\; f_w(x_i) = y_i $$
we want to find a function, $f$, such that $f(x_i) \approx y_i$ for all $i$. This process is called function approximation. In data science, we often use linear functions to approximate complex relationships between inputs and outputs, which is the basis of linear regression and other linear models.

### How does linear algebra help?
Linear functions can well approximate the relationships between most underlying objects. Even if the actual relationship isn’t linear, it can be approximated through several linear functions. This is precisely what happens in deep neural networks too. Linearity makes many things tractable and straightforward. Linearity is useful in function approximation, particularly in data science.
