# Convex Combinations

A convex combination of a set of points is a linear combination of those points where all coefficients are non-negative and sum to one. A convex combination makes a result that stays between the original objects, instead of going outside them.

A convex combination of objects $x_1, x_2, \dots, x_n$ is $\alpha_1 x_1 + \alpha_2 x_2 + \cdots + \alpha_n x_n$

with:

$$
\alpha_i \ge 0
\quad \text{and} \quad
\sum_i \alpha_i = 1
$$


This means:
- each weight is non-negative, so we are not “subtracting” one object from another
- all weights add to 1, so the result is a true weighted average

A convex combination is a weighted average of vectors, matrices, probabilities, or features And weighted averages are everywhere in ML.

This concept is important in machine learning because it allows us to combine information in a stable and meaningful way. Since the coefficients are non-negative and sum to one, the resulting object stays between the original objects instead of going far outside them. This makes convex combinations safer and more interpretable than arbitrary linear combinations.



## Why do we use convex combinations?

### keeps the output within a reasonable range
It keeps the output within a reasonable range. If you average two vectors, two probability distributions, or two images using convex weights, the result stays between them.

For example:

$0.3v_1 + 0.7v_2$ lies between $v_1$ and $v_2$.

This is useful because in ML we often want a result that is:
- stable
- bounded
- not extreme
- interpretable

Because a convex combination is a weighted average, it cannot go outside the values of the things you are averaging.

If

$$
z = \alpha_1 x_1 + \alpha_2 x_2 + \cdots + \alpha_n x_n
$$
with

$$
\alpha_i \ge 0, \quad \sum_i \alpha_i = 1
$$

then the output stays “between” the inputs.

#### Simple number example

Suppose we have two values:

$$
x_1 = 10,\quad x_2 = 20
$$

Now take a convex combination:

$$
z = 0.3(10) + 0.7(20) \\
z = 3 + 14 = 17
$$

Notice:
- output is not less than 10
- output is not greater than 20
- it stays between them

If we change the weights:

$$
z = 0.8(10) + 0.2(20) = 12
$$

Still between 10 and 20. So for plain numbers, a convex combination always stays inside the minimum and maximum.


#### Why this happens
Because the weights are:
- non-negative
- sum to 1

they behave like proportions.

$$
0.3x_1 + 0.7x_2
$$

means:
- 30% of the first value
- 70% of the second value

That is just an average with unequal importance. And averages do not jump outside the original range.


#### What goes wrong with a normal linear combination
Now compare with:

$$
z = 2(10) - 1(20) = 0
$$
This is a linear combination, but not a convex combination.

The output is 0, which is outside the range [10, 20].

So ordinary linear combinations can produce extreme values. Convex combinations avoid that.


#### Vector example

Let

$$
v_1 =
\begin{bmatrix}
2 \\
4
\end{bmatrix},
\quad
v_2 =
\begin{bmatrix}
6 \\
10
\end{bmatrix}
$$

Take a convex combination:

$$
z = 0.25v_1 + 0.75v_2 \\
z = 0.25 \\

\begin{bmatrix}
2 \\
4
\end{bmatrix}
+
0.75
\begin{bmatrix}
6 \\
10
\end{bmatrix}

=
\begin{bmatrix}
0.5 \\
1
\end{bmatrix}
+
\begin{bmatrix}
4.5 \\
7.5
\end{bmatrix}
=
\begin{bmatrix}
5 \\
8.5
\end{bmatrix}
$$


Look at each coordinate:
- first coordinate: between 2 and 6
- second coordinate: between 4 and 10

So the new vector stays between the original vectors coordinate-wise. Geometrically, it lies on the line segment joining $v_1$ and $v_2$.

5. Probability example

Suppose two class-probability vectors are:

p_1 = [0.9, 0.1]

p_2 = [0.2, 0.8]

Now mix them:

p = 0.6p_1 + 0.4p_2

p = 0.6[0.9,0.1] + 0.4[0.2,0.8]

= [0.54, 0.06] + [0.08, 0.32]

= [0.62, 0.38]

Why is this good?
- each value is still between 0 and 1
- entries are still non-negative
- total still sums to 1

So the result is still a valid probability distribution. This is one of the biggest reasons convex combinations are useful in ML.


8. The geometric meaning

For two points x_1 and x_2, every convex combination

$$
(1-\alpha)x_1 + \alpha x_2, \quad 0 \le \alpha \le 1
$$

lies exactly on the line segment between them. So it cannot go beyond the endpoints.
For many points, the result stays inside their convex hull, which means inside the smallest “shape” containing all those points.

That is the deeper reason it stays within a reasonable range.



### It represents “soft mixing” or “blending”

Many ML systems do not want to fully choose one thing. Instead, they want to combine several possibilities.

Convex combinations let us say:
 - 70% from one source
 - 20% from another
 - 10% from another

Examples:
 - combining multiple model predictions
 - mixing embeddings
 - attention over tokens
 - interpolating between latent vectors
 - smoothing class probabilities

### It preserves probability meaning

Probabilities must be:
 - non-negative
 - sum to 1

That is exactly the same structure as convex coefficients. So whenever ML works with probabilities, convex combinations naturally appear.

Suppose one model predicts:

$$
p_1 = [0.8, 0.2]
$$

and another predicts:

$$
p_2 = [0.4, 0.6]
$$

A convex combination gives:

$$
0.75p_1 + 0.25p_2 = [0.7, 0.3]
$$

This is still a valid probability distribution. That is why convex combinations are used in:
- ensemble learning
- label smoothing
- Bayesian model averaging
- mixture models
- attention weights
- soft targets in distillation


### It makes optimization more well-behaved

Convexity is one of the nicest properties in optimization.

When we work with convex combinations:
 - the search space is more controlled
 - interpolation between solutions is meaningful
 - many mathematical guarantees become easier

In ML, training often involves optimization, so convex ideas are powerful because they reduce instability and make analysis easier. Even when deep learning is not fully convex, convex combinations still appear inside many parts of the system because they are numerically and conceptually useful.


5. It creates smooth transitions

Your image morphing example is a perfect illustration.

$$
I_\alpha = (1-\alpha)A + \alpha B
$$

When $\alpha$ moves from 0 to 1:
- at $\alpha = 0$, the result is all A
- at $\alpha = 1$, the result is all B
- in between, the result is a smooth blend

This same idea appears in ML when we want smooth interpolation between representations.

For example:
 - interpolating between embeddings
 - latent space traversal in generative models
 - data augmentation like mixup
 - transition between two policies in reinforcement learning


## Why not just use any linear combination?

Because a general linear combination can do things we may not want.

$$
1.4x_1 - 0.4x_2
$$

This is a valid linear combination, but not a convex combination.

Why is it risky?
- one weight is negative
- weights do not represent proportions cleanly
- the result can move outside the region spanned “between” the points
- it may produce unrealistic values

So:
- linear combination = flexible
- convex combination = controlled and safe

## Convex combinations in machine learning

1. Attention mechanism

In transformers, the output for a token is a weighted sum of value vectors:

$$
\text{output} = \sum_i \alpha_i v_i
$$

where attention weights satisfy:

$$
\alpha_i \ge 0, \quad \sum_i \alpha_i = 1
$$

So attention output is a convex combination of value vectors. Why is this useful?
Because the model can softly focus on multiple relevant tokens while keeping the representation stable.


2. Ensemble models

If two models make predictions, we may combine them as:

$$
\hat{y} = 0.6\hat{y}_1 + 0.4\hat{y}_2
$$

This is a convex combination. Why?
Because we want a balanced final prediction, not one that becomes too extreme.


3. Mixup data augmentation

In mixup, we create synthetic training examples like:

$$
\tilde{x} = \lambda x_i + (1-\lambda)x_j
$$

and labels:

$$
\tilde{y} = \lambda y_i + (1-\lambda)y_j
$$
This is a convex combination of both inputs and labels. Why does this help? Because it encourages smoother decision boundaries and reduces overfitting.


4. Probability distributions

If you combine two distributions using convex weights, the result is still a valid distribution.

This is used in:
- mixture models
- label smoothing
- distillation
- uncertainty estimation


5. Embedding interpolation

In representation learning, convex combinations can blend meanings or features while staying in a more natural region of space. That is why interpolations often look smoother and more realistic than arbitrary extrapolations.
