# Linear Independence

 Linear independence is one of the most important ideas in linear algebra for machine learning and deep learning, because it tells us whether our features, vectors, or representations carry new information or just repeat information we already have.

 ## Core intuition

    A set of vectors is linearly independent if no vector can be written as a linear combination of the others.
    In other words, each vector adds something new that cannot be formed by mixing the others.
    If a vector can be formed by mixing the others, it is linearly dependent, meaning it doesn't add new information.

Suppose you have several features, vectors, or neurons. If one of them can be written as a combination of the others, then it is redundant.
It is not adding truly new information.

That is exactly what linear dependence means.

If a vector cannot be created from the others, then it contributes a new direction or new information.
That is what linear independence means.

So in machine learning:
- linearly independent features = more unique information
- linearly dependent features = repeated or redundant information


We use the idea of linear independence in ML and DL because it helps answer questions like:
- Are my input features actually different from each other?
- Am I giving the model repeated information?
- Is my representation space rich enough?
- Are some dimensions useless or redundant?
- Can this data span the space needed to solve the problem?
- Why is matrix inversion failing or unstable?
- Why do some embeddings or learned features collapse?

### Example
Simple intuition with features. Suppose your dataset has these features:
- height in centimeters
- height in meters

These are clearly dependent, because:

$$
\text{height in meters} = \frac{1}{100} \times \text{height in centimeters}
$$

So one feature adds no new information. If you include both, the model is seeing duplicate information in a different form.
Now suppose you also add:
- age
- number of purchases
- average session time

These may provide genuinely new information, so they are more likely to be independent. This is why linear independence matters: it helps us think about whether features are meaningful or redundant.


Suppose we have two vectors:

$$
v_1 = \begin{bmatrix}2 \\
4
\end{bmatrix}, \quad v_2 = \begin{bmatrix}6 \\
10
\end{bmatrix}
$$
Can we write $v_2$ as a linear combination of $v_1$?
Yes, we can:

$$
v_2 = 3v_1
$$
This means $v_2$ is linearly dependent on $v_1$. It does not add new information. Both vectors lie on the same line in 2D space.

Now, if we take a convex combination of $v_1$ and $v_2$:

$$
\tilde{v} = \lambda v_1 + (1-\lambda)v_2, \quad 0 \le \lambda \le 1
$$
This new vector $\tilde{v}$ will lie on the line segment between $v_1$ and $v_2$. It will not go beyond them, because both $v_1$ and $v_2$ are on the same line. So $\tilde{v}$ is also linearly dependent on $v_1$ (and $v_2$).



## Importance

### Detect Redundant Features
Linear dependence can help us detect redundant features. If we find that one feature can be written as a combination of others, we can remove it without losing information. This can simplify our model and reduce overfitting.

This is one of the most practical reasons. If some features are linearly dependent, then one feature can be exactly predicted from others.

That means:
- the feature set has redundancy
- the model may become unnecessarily complex
- some parameters may become hard to interpret

If:

$$
x_3 = 2x_1 + 5x_2
$$

then $x_3$ does not add new information beyond $x_1$ and $x_2$.

In ML, redundant features can hurt:
- interpretability
- efficiency
- numerical stability


### whether a matrix is invertible
Linear independence is crucial for determining whether a matrix is invertible. Many ML methods use matrices. If the columns of a matrix are linearly dependent, then the matrix does not have full rank. That can cause serious issues. For example, in linear regression, the normal equation uses:

$$
(X^TX)^{-1}
$$

If the columns of X are linearly dependent, then $X^TX$ may be singular or non-invertible.

This means:
- the model may not have a unique solution
- optimization can become unstable
- parameter estimates can explode or become unreliable

So linear independence is critical for stable learning.

### It helps ensure each learned direction is useful

In representation learning, embeddings, hidden states, or latent vectors should ideally capture different aspects of the data. If many learned vectors are dependent, then the model may be wasting dimensions.

Instead of learning diverse information, it keeps repeating the same direction. This can reduce the expressive power of the representation.

In deep learning terms, if hidden units all behave similarly, the network is not using its capacity well.

So linear independence is connected to:
- diversity of learned features
- non-collapse of representations
- efficient use of embedding dimensions


### Explains dimensionality reduction

Methods like PCA are deeply connected to linear independence. Why do we do PCA? 
Because real data often lives in a lower-dimensional structure, even if it is stored in a high-dimensional space.

For example:
- 100 features may actually contain only 10 meaningful independent directions
- the remaining dimensions may be combinations or noisy variants of those directions

PCA tries to find the most important independent directions of variation.

So linear independence helps explain:
- why data can be compressed
- why some dimensions are unnecessary
- why latent structure exists


### Improves numerical stability

Dependent or nearly dependent features create multicollinearity. This can make optimization unstable.
Small changes in data can lead to large changes in learned weights. So even if the model still trains, the learned parameters can become unreliable.

This matters in:
- linear regression
- logistic regression
- generalized linear models
- some neural network settings
- covariance estimation
- kernel methods

So linear independence matters not only mathematically but computationally.


### Determines whether vectors span the space

A set of vectors can only represent a space well if enough of them are independent.

For example:
- in 2D, two independent vectors can span the plane
- in 3D, three independent vectors can span the full space
- if vectors are dependent, they span only a smaller region

This matters in ML because models often represent data in vector spaces. If learned vectors do not span enough useful directions, the model cannot express the needed patterns.


## Connect linear independence directly to common ML topics

### Linear regression

In linear regression, we use input matrix X, where columns are features. If some columns are linearly dependent, then:
- the model parameters are not uniquely identifiable
- coefficients can become unstable
- the inverse in the closed-form solution may not exist


Suppose:
- $x_1$ = house size in square feet
- $x_2$ = house size in square meters

These features are dependent. Then the regression model may struggle to assign a unique meaningful weight to each one.

It can still predict, but interpretation becomes messy.

Why?

Because many different weight combinations can produce the same output.

So linear independence is important for:
- unique solutions
- stable coefficients
- better interpretability


### PCA and feature extraction

PCA looks for directions that explain the variance in the data. If features are dependent, then the true number of useful dimensions is smaller than the raw number of columns.

Linear independence tells us:
- how many meaningful directions exist
- how much of the space is redundant
- how much compression is possible

So PCA is strongly tied to rank and independence.


### Embeddings

Word embeddings, item embeddings, and latent vectors should ideally capture different semantic or behavioral directions.
If many embedding dimensions collapse into similar directions, then the embedding space is less expressive.

Linear independence helps us understand:
- diversity in embeddings
- useful dimensionality
- why some learned spaces are richer than others

For example, if every dimension in an embedding table carried the same information, that embedding would be weak.


### Neural network hidden representations

In deep learning, hidden layers transform input data into learned representations. If neurons produce highly redundant outputs, then the layer is not learning diverse features.

That means:
- the representation is inefficient
- capacity is underused
- information may collapse into fewer effective dimensions

This is related to ideas like:
- rank collapse
- feature redundancy
- dead neurons
- poor representation diversity

So linear independence helps us think about whether a deep model is learning useful and distinct features.




## Formal idea in simpler language

A set of vectors $x_1, x_2, \dots, x_n$ is linearly dependent if there exist scalars $w_1, w_2, \dots, w_n$, not all zero, such that

$$
w_1x_1 + w_2x_2 + \cdots + w_nx_n = 0
$$

This means one vector in the set can be expressed using the others. So the set contains redundancy. If the only solution is all zeros, then the vectors are linearly independent.

## Test for linear independence

we check whether any vector in a set can be built from the others.

A set of vectors $v_1, v_2, \dots, v_n$ is linearly independent if the equation

$$
c_1v_1 + c_2v_2 + \cdots + c_nv_n = 0
$$

has only the trivial solution:

$$
c_1 = c_2 = \cdots = c_n = 0
$$

If there is any non-zero choice of coefficients that makes the sum equal to the zero vector, then the vectors are linearly dependent.

Can these vectors cancel each other out to make the zero vector without using all-zero coefficients?

- Only all zeros work → independent
- Some non-zero coefficients work → dependent

### Why this works

If one vector can be written as a combination of the others, then the set contains redundancy. For example, if

$$
v_3 = 2v_1 + v_2
$$

then move everything to one side:

$$
2v_1 + v_2 - v_3 = 0
$$

Now we found a non-zero combination equal to zero, so the vectors are dependent. That is the reason this test works.



### Example 1: Two independent vectors

Test whether

$$

v_1 = \begin{bmatrix}1\\0\end{bmatrix},
\quad
v_2 = \begin{bmatrix}0\\1\end{bmatrix}
$$

are linearly independent.

We write:

$$

c_1v_1 + c_2v_2 = 0
$$

Substitute the vectors:

$$
c_1\begin{bmatrix}1\\0\end{bmatrix}
+
c_2\begin{bmatrix}0\\1\end{bmatrix}
=
\begin{bmatrix}0\\0\end{bmatrix}
$$

This gives:

$$

\begin{bmatrix}c_1\\c_2\end{bmatrix}
=
\begin{bmatrix}0\\0\end{bmatrix}
$$

So:

$$
c_1 = 0,\quad c_2 = 0
$$

Only the trivial solution exists.

So the vectors are linearly independent.

Reason

Neither vector can be made from the other.
Each gives a new direction.

### Example 2: Two dependent vectors

Test whether

$$

v_1 = \begin{bmatrix}1\\2\end{bmatrix},
\quad
v_2 = \begin{bmatrix}2\\4\end{bmatrix}

$$

are linearly independent.

Write:

$$

c_1v_1 + c_2v_2 = 0
$$

Substitute:

$$

c_1\begin{bmatrix}1\\2\end{bmatrix}
+
c_2\begin{bmatrix}2\\4\end{bmatrix}
=
\begin{bmatrix}0\\0\end{bmatrix}

$$

This gives the system:

$$

c_1 + 2c_2 = 0
2c_1 + 4c_2 = 0
$$

The second equation is just twice the first, so there are infinitely many solutions.

For example:

$$

c_2 = 1,\quad c_1 = -2
$$

Then:

$$

-2v_1 + 1v_2 = 0
$$

So the vectors are linearly dependent.

Reason

v_2 = 2v_1, so one vector is just a scaled copy of the other.








Method 2: Put vectors into a matrix and row reduce

This is the most standard method.

Put the vectors as columns of a matrix, then perform Gaussian elimination.
	•	If every column has a pivot, the vectors are independent.
	•	If a column does not get a pivot, the vectors are dependent.

⸻

Example 3: Use row reduction

Test whether these vectors are independent:

v_1=\begin{bmatrix}1\\0\\1\end{bmatrix},
\quad
v_2=\begin{bmatrix}0\\1\\1\end{bmatrix},
\quad
v_3=\begin{bmatrix}1\\1\\2\end{bmatrix}

Make a matrix with these as columns:

A=
\begin{bmatrix}
1 & 0 & 1 \\
0 & 1 & 1 \\
1 & 1 & 2
\end{bmatrix}

Now row reduce.

Start with:

\begin{bmatrix}
1 & 0 & 1 \\
0 & 1 & 1 \\
1 & 1 & 2
\end{bmatrix}

Do:

R_3 \leftarrow R_3 - R_1

\begin{bmatrix}
1 & 0 & 1 \\
0 & 1 & 1 \\
0 & 1 & 1
\end{bmatrix}

Now:

R_3 \leftarrow R_3 - R_2

\begin{bmatrix}
1 & 0 & 1 \\
0 & 1 & 1 \\
0 & 0 & 0
\end{bmatrix}

There are only 2 pivots, not 3.

So the vectors are linearly dependent.

Reason

One vector can be made from the others. In fact,

v_3 = v_1 + v_2

So the third vector adds no new direction.

⸻

Example 4: Independent 3D vectors

Test:

v_1=\begin{bmatrix}1\\0\\0\end{bmatrix},
\quad
v_2=\begin{bmatrix}0\\1\\0\end{bmatrix},
\quad
v_3=\begin{bmatrix}0\\0\\1\end{bmatrix}

Matrix:

A=
\begin{bmatrix}
1 & 0 & 0 \\
0 & 1 & 0 \\
0 & 0 & 1
\end{bmatrix}

This already has a pivot in every column.

So the vectors are linearly independent.

Reason

These are the standard basis vectors in 3D.
Each points in a unique axis direction.

Method 3: Determinant test

This method works only when the number of vectors equals the dimension, so the matrix is square.

If the determinant is:
	•	non-zero → vectors are independent
	•	zero → vectors are dependent

⸻

Example 5: Determinant non-zero

Test:

v_1=\begin{bmatrix}1\\2\end{bmatrix},
\quad
v_2=\begin{bmatrix}3\\4\end{bmatrix}

Matrix:

A=
\begin{bmatrix}
1 & 3 \\
2 & 4
\end{bmatrix}

Determinant:

\det(A)= (1)(4) - (3)(2) = 4-6 = -2

Since determinant is non-zero, the vectors are linearly independent.

Reason

A non-zero determinant means the matrix is invertible, so its columns are independent.

⸻

Example 6: Determinant zero

Test:

v_1=\begin{bmatrix}1\\2\end{bmatrix},
\quad
v_2=\begin{bmatrix}2\\4\end{bmatrix}

Matrix:

A=
\begin{bmatrix}
1 & 2 \\
2 & 4
\end{bmatrix}

Determinant:

\det(A)= (1)(4) - (2)(2)=4-4=0

So the vectors are linearly dependent.

Reason

The second column is twice the first.

Quick tests and shortcuts

These are useful for fast checking.

⸻

1. A single non-zero vector is always independent

Example:

v=\begin{bmatrix}3\\5\end{bmatrix}

The only way

cv = 0

can happen is if c=0.

So it is independent.

Reason

A non-zero vector cannot cancel itself unless multiplied by zero.

⸻

2. Any set containing the zero vector is dependent

Example:

v_1=\begin{bmatrix}1\\2\end{bmatrix},
\quad
v_2=\begin{bmatrix}0\\0\end{bmatrix}

Take:

0v_1 + 5v_2 = 0

This uses a non-zero coefficient, so the set is dependent.

Reason

The zero vector immediately creates a non-trivial solution.

⸻

3. More vectors than dimension means dependent

In \mathbb{R}^2, any set of 3 vectors must be dependent.
In \mathbb{R}^3, any set of 4 vectors must be dependent.

Example in \mathbb{R}^2:

\begin{bmatrix}1\\0\end{bmatrix},
\quad
\begin{bmatrix}0\\1\end{bmatrix},
\quad
\begin{bmatrix}1\\1\end{bmatrix}

These are dependent because:

\begin{bmatrix}1\\1\end{bmatrix}
=
\begin{bmatrix}1\\0\end{bmatrix}
+
\begin{bmatrix}0\\1\end{bmatrix}

Reason

You cannot have more than 2 independent directions in 2D.

⸻

4. Two vectors are dependent if one is a scalar multiple of the other

Example:

\begin{bmatrix}2\\3\end{bmatrix},
\quad
\begin{bmatrix}4\\6\end{bmatrix}

Since

\begin{bmatrix}4\\6\end{bmatrix}
=
2\begin{bmatrix}2\\3\end{bmatrix}

they are dependent.

Reason

They lie on the same line and give the same direction.

Geometric meaning

This helps a lot.

In 2D

Two vectors are:
	•	independent if they point in different directions and span the plane
	•	dependent if they lie on the same line

In 3D

Three vectors are:
	•	independent if they span the whole 3D space
	•	dependent if one lies in the plane formed by the other two

Why this matters

Testing for linear independence tells us whether vectors bring new information.

In ML terms:
	•	independent features = more unique information
	•	dependent features = redundancy

That is why rank, invertibility, PCA, and feature selection all connect to linear independence.

⸻

Best practical procedure

If you are solving homework or interview-style questions, use this order:

For 2 vectors

Check whether one is a scalar multiple of the other.

For square matrices

Use determinant.

For general sets of vectors

Put them into a matrix and row reduce.

That is the safest universal method.

Final summary

A set of vectors is linearly independent if the only solution to

c_1v_1 + c_2v_2 + \cdots + c_nv_n = 0

is

c_1=c_2=\cdots=c_n=0

It is linearly dependent if there is any non-zero solution.

Main ways to test:
	•	direct equation solving
	•	row reduction
	•	determinant for square cases
	•	quick checks like zero vector, scalar multiples, or too many vectors

Main reason:

Because dependence means redundancy, while independence means each vector contributes a new direction.
