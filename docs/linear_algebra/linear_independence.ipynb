# Linear Independence

 Linear independence is one of the most important ideas in linear algebra for machine learning and deep learning, because it tells us whether our features, vectors, or representations carry new information or just repeat information we already have.

 ## Core intuition

**A set of vectors is linearly independent if no vector can be written as a linear combination of the others.
In other words, each vector adds something new that cannot be formed by mixing the others.
If a vector can be formed by mixing the others, it is linearly dependent, meaning it doesn't add new information.**

Suppose you have several features, vectors, or neurons. If one of them can be written as a combination of the others, then it is redundant.
It is not adding truly new information.

That is exactly what linear dependence means.

If a vector cannot be created from the others, then it contributes a new direction or new information.
That is what linear independence means.

So in machine learning:
- linearly independent features = more unique information
- linearly dependent features = repeated or redundant information


We use the idea of linear independence in ML and DL because it helps answer questions like:
- Are my input features actually different from each other?
- Am I giving the model repeated information?
- Is my representation space rich enough?
- Are some dimensions useless or redundant?
- Can this data span the space needed to solve the problem?
- Why is matrix inversion failing or unstable?
- Why do some embeddings or learned features collapse?

### Example
Simple intuition with features. Suppose your dataset has these features:
- height in centimeters
- height in meters

These are clearly dependent, because:

$$
\text{height in meters} = \frac{1}{100} \times \text{height in centimeters}
$$

So one feature adds no new information. If you include both, the model is seeing duplicate information in a different form.
Now suppose you also add:
- age
- number of purchases
- average session time

These may provide genuinely new information, so they are more likely to be independent. This is why linear independence matters: it helps us think about whether features are meaningful or redundant.


Suppose we have two vectors:

$$
v_1 = \begin{bmatrix}2 \\
4
\end{bmatrix}, \quad v_2 = \begin{bmatrix}6 \\
10
\end{bmatrix}
$$
Can we write $v_2$ as a linear combination of $v_1$?
Yes, we can:

$$
v_2 = 3v_1
$$
This means $v_2$ is linearly dependent on $v_1$. It does not add new information. Both vectors lie on the same line in 2D space.

Now, if we take a convex combination of $v_1$ and $v_2$:

$$
\tilde{v} = \lambda v_1 + (1-\lambda)v_2, \quad 0 \le \lambda \le 1
$$
This new vector $\tilde{v}$ will lie on the line segment between $v_1$ and $v_2$. It will not go beyond them, because both $v_1$ and $v_2$ are on the same line. So $\tilde{v}$ is also linearly dependent on $v_1$ (and $v_2$).



## Importance

### Detect Redundant Features
Linear dependence can help us detect redundant features. If we find that one feature can be written as a combination of others, we can remove it without losing information. This can simplify our model and reduce overfitting.

This is one of the most practical reasons. If some features are linearly dependent, then one feature can be exactly predicted from others.

That means:
- the feature set has redundancy
- the model may become unnecessarily complex
- some parameters may become hard to interpret

If:

$$
x_3 = 2x_1 + 5x_2
$$

then $x_3$ does not add new information beyond $x_1$ and $x_2$.

In ML, redundant features can hurt:
- interpretability
- efficiency
- numerical stability


### whether a matrix is invertible
Linear independence is crucial for determining whether a matrix is invertible. Many ML methods use matrices. If the columns of a matrix are linearly dependent, then the matrix does not have full rank. That can cause serious issues. For example, in linear regression, the normal equation uses:

$$
(X^TX)^{-1}
$$

If the columns of X are linearly dependent, then $X^TX$ may be singular or non-invertible.

This means:
- the model may not have a unique solution
- optimization can become unstable
- parameter estimates can explode or become unreliable

So linear independence is critical for stable learning.

### It helps ensure each learned direction is useful

In representation learning, embeddings, hidden states, or latent vectors should ideally capture different aspects of the data. If many learned vectors are dependent, then the model may be wasting dimensions.

Instead of learning diverse information, it keeps repeating the same direction. This can reduce the expressive power of the representation.

In deep learning terms, if hidden units all behave similarly, the network is not using its capacity well.

So linear independence is connected to:
- diversity of learned features
- non-collapse of representations
- efficient use of embedding dimensions


### Explains dimensionality reduction

Methods like PCA are deeply connected to linear independence. Why do we do PCA? 
Because real data often lives in a lower-dimensional structure, even if it is stored in a high-dimensional space.

For example:
- 100 features may actually contain only 10 meaningful independent directions
- the remaining dimensions may be combinations or noisy variants of those directions

PCA tries to find the most important independent directions of variation.

So linear independence helps explain:
- why data can be compressed
- why some dimensions are unnecessary
- why latent structure exists


### Improves numerical stability

Dependent or nearly dependent features create multicollinearity. This can make optimization unstable.
Small changes in data can lead to large changes in learned weights. So even if the model still trains, the learned parameters can become unreliable.

This matters in:
- linear regression
- logistic regression
- generalized linear models
- some neural network settings
- covariance estimation
- kernel methods

So linear independence matters not only mathematically but computationally.


### Determines whether vectors span the space

A set of vectors can only represent a space well if enough of them are independent.

For example:
- in 2D, two independent vectors can span the plane
- in 3D, three independent vectors can span the full space
- if vectors are dependent, they span only a smaller region

This matters in ML because models often represent data in vector spaces. If learned vectors do not span enough useful directions, the model cannot express the needed patterns.


## Connect linear independence directly to common ML topics

### Linear regression

In linear regression, we use input matrix X, where columns are features. If some columns are linearly dependent, then:
- the model parameters are not uniquely identifiable
- coefficients can become unstable
- the inverse in the closed-form solution may not exist


Suppose:
- $x_1$ = house size in square feet
- $x_2$ = house size in square meters

These features are dependent. Then the regression model may struggle to assign a unique meaningful weight to each one.

It can still predict, but interpretation becomes messy.

Why?

Because many different weight combinations can produce the same output.

So linear independence is important for:
- unique solutions
- stable coefficients
- better interpretability


### PCA and feature extraction

PCA looks for directions that explain the variance in the data. If features are dependent, then the true number of useful dimensions is smaller than the raw number of columns.

Linear independence tells us:
- how many meaningful directions exist
- how much of the space is redundant
- how much compression is possible

So PCA is strongly tied to rank and independence.


### Embeddings

Word embeddings, item embeddings, and latent vectors should ideally capture different semantic or behavioral directions.
If many embedding dimensions collapse into similar directions, then the embedding space is less expressive.

Linear independence helps us understand:
- diversity in embeddings
- useful dimensionality
- why some learned spaces are richer than others

For example, if every dimension in an embedding table carried the same information, that embedding would be weak.


### Neural network hidden representations

In deep learning, hidden layers transform input data into learned representations. If neurons produce highly redundant outputs, then the layer is not learning diverse features.

That means:
- the representation is inefficient
- capacity is underused
- information may collapse into fewer effective dimensions

This is related to ideas like:
- rank collapse
- feature redundancy
- dead neurons
- poor representation diversity

So linear independence helps us think about whether a deep model is learning useful and distinct features.




## Formal idea in simpler language

A set of vectors $x_1, x_2, \dots, x_n$ is linearly dependent if there exist scalars $w_1, w_2, \dots, w_n$, not all zero, such that

$$
w_1x_1 + w_2x_2 + \cdots + w_nx_n = 0
$$

This means one vector in the set can be expressed using the others. So the set contains redundancy. If the only solution is all zeros, then the vectors are linearly independent.

## Test for linear independence

The real meaning of linear independence, Suppose you have some vectors.

You want to know:

“Does each vector add a new direction, or is one of them just made from the others?”

That is the whole question.

- If every vector adds a new direction, the set is linearly independent
- If at least one vector can be built from the others, the set is linearly dependent

So the test is really a test for redundancy.


The one idea you must remember forever

A set of vectors

$$
v_1, v_2, \dots, v_n
$$

is linearly independent if the equation

$$
c_1v_1 + c_2v_2 + \cdots + c_nv_n = 0
$$

has only one solution:

$$

c_1 = c_2 = \cdots = c_n = 0
$$

This is called the trivial solution. If there is any other solution where at least one coefficient is non-zero, then the vectors are dependent.

Why do we test using “equals zero”? Why do we test

$$
c_1v_1 + c_2v_2 + \cdots + c_nv_n = 0
$$

and not something else? Because this equation reveals whether one vector can be written using the others.

For example, suppose

$$
v_3 = 2v_1 + 5v_2
$$

Then we can rearrange this to:

$$

2v_1 + 5v_2 - v_3 = 0
$$

Now you found a combination equal to zero, and the coefficients are not all zero. That means the vectors are dependent.
So the “zero test” is the universal way to detect whether one vector is secretly built from the others.
That is why it works.


### Geometric intuition

In 2D, two vectors are:

- independent if they point in different directions
- dependent if they lie on the same line

Why?

Because if they lie on the same line, one is just a multiple of the other.

$$

\begin{bmatrix}2\\4\end{bmatrix}
=
2
\begin{bmatrix}1\\2\end{bmatrix}

$$

So these are dependent.

But

$$

\begin{bmatrix}1\\0\end{bmatrix},
\quad
\begin{bmatrix}0\\1\end{bmatrix}
$$

point in different directions, so together they cover the plane.

These are independent.  

In 3D, three vectors are independent if they do not lie in the same plane. If they lie in the same plane, one can be formed from the others, so they are dependent.

Example: in 3D, the vectors

$$
\begin{bmatrix}1\\0\\0\end{bmatrix},
\quad
\begin{bmatrix}0\\1\\0\end{bmatrix},
\quad
\begin{bmatrix}0\\0\\1\end{bmatrix}
$$

These vectors are independent because they do not lie in the same plane.

In general, in n-dimensional space, n vectors are independent if they do not lie in the same (n-1)-dimensional hyperplane. If they lie in the same hyperplane, then one can be formed from the others, so they are dependent.


### The universal method

This is the method you should learn first because it always works.

Step 1

Write the vectors as columns of a matrix.

Step 2

Solve

$$

A\mathbf{c}=0
$$

where A is the matrix of vectors and $\mathbf{c}$ is the coefficient vector.

Step 3

Row reduce the matrix.

Step 4

Look at the solution.

- If the only solution is all zeros, the vectors are independent
- If there is a free variable, then there are infinitely many non-zero solutions, so the vectors are dependent

Why row reduction works Because row reduction does not change the solution set of the system.

So when you reduce the matrix, you are just making the equation easier to solve.

You are still solving the same independence question:

$$
c_1v_1 + c_2v_2 + \cdots + c_nv_n = 0
$$



Example 1: two vectors in 2D, independent

$$
v_1=
\begin{bmatrix}
1\\0
\end{bmatrix},
\quad
v_2=
\begin{bmatrix}
0\\1
\end{bmatrix}
$$

These are the standard basis vectors.

Step 1: Make matrix

$$
A=
\begin{bmatrix}
1 & 0\\
0 & 1
\end{bmatrix}
$$

Step 2: Solve

$$
A
\begin{bmatrix}
c_1\\
c_2
\end{bmatrix}
=
\begin{bmatrix}
0\\
0
\end{bmatrix}
$$

So:

$$

\begin{bmatrix}
1 & 0\\
0 & 1
\end{bmatrix}
\begin{bmatrix}
c_1\\
c_2
\end{bmatrix}
=
\begin{bmatrix}
0\\
0
\end{bmatrix}
$$

This gives:

$$

c_1 = 0,\quad c_2 = 0
$$

Only trivial solution.

So the vectors are independent.

Why this example works

Because neither vector can be made from the other.

- v_1 moves horizontally
- v_2 moves vertically

These are genuinely different directions.

So together they give new information.

Example 2: two vectors in 2D, dependent

Test:

$$

v_1=
\begin{bmatrix}
1\\2
\end{bmatrix},
\quad
v_2=
\begin{bmatrix}
2\\4
\end{bmatrix}
$$
First quick observation

$$

v_2 = 2v_1

$$

So already you should suspect dependence.

Step 1: Make matrix

$$

A=
\begin{bmatrix}
1 & 2\\
2 & 4
\end{bmatrix}
$$

Step 2: Solve

$$

A
\begin{bmatrix}
c_1\\
c_2
\end{bmatrix}
=
\begin{bmatrix}
0\\
0
\end{bmatrix}
$$

This means:

$$

\begin{cases}
c_1 + 2c_2 = 0\\
2c_1 + 4c_2 = 0
\end{cases}
$$

But the second equation is just twice the first.

So really there is only one independent equation.

From

$$

c_1 + 2c_2 = 0
$$

we get

$$

c_1 = -2c_2
$$

Now choose any value for $c_2$.

Let $c_2=1$, then

$c_1=-2$

So

$$
-2v_1 + 1v_2 = 0
$$

This is a non-trivial solution.

So the vectors are dependent.

Why this example works

Because both vectors point in exactly the same direction.

The second is just a stretched copy of the first.

So the second vector does not add a new direction.

It is redundant.


The biggest memory trick for 2 vectors

For two vectors only, the test is very easy:

Two vectors are dependent if one is a scalar multiple of the other.

That means:

$v_2 = kv_1$

for some scalar k.

If yes, dependent.
If no, independent.

This is the fastest test for two vectors.


Example 3: three vectors in 2D

Test:

$$

v_1=
\begin{bmatrix}
1\\0
\end{bmatrix},
\quad
v_2=
\begin{bmatrix}
0\\1
\end{bmatrix},
\quad
v_3=
\begin{bmatrix}
1\\1
\end{bmatrix}
$$

Before doing any algebra

These are three vectors in 2D.

In 2D, you can have at most two independent vectors.

Why?

Because 2D space has only two independent directions.

So any set of 3 vectors in 2D must be dependent.

Direct reason

$$

v_3 = v_1 + v_2
$$

So

$$

v_1 + v_2 - v_3 = 0
$$

Non-trivial solution.

So dependent.

Why this example works

Because once you already have horizontal and vertical directions, any other 2D vector can be built from them.

So the third vector cannot be truly new.

This is a very important fact:

In $\mathbb{R}^n$, more than $n$ vectors must be dependent.

Example 4: three vectors in 3D, dependent

Test:

$$
v_1=
\begin{bmatrix}
1\\0\\1
\end{bmatrix},
\quad
v_2=
\begin{bmatrix}
0\\1\\1
\end{bmatrix},
\quad
v_3=
\begin{bmatrix}
1\\1\\2
\end{bmatrix}
$$

Try to notice a pattern

Add $v_1$ and $v_2$:

$$
v_1 + v_2
=
\begin{bmatrix}
1\\0\\1
\end{bmatrix}
+
\begin{bmatrix}

\end{bmatrix}
=
\begin{bmatrix}

\end{bmatrix}
=

$$

So

$$

v_1 + v_2 - v_3 = 0
$$

Non-trivial solution.

Therefore dependent.

Why this example works

Because the third vector is already made from the first two.

So the third adds nothing new.

This is exactly what dependence means.


The matrix method in full detail

Now let’s do the proper systematic method.

Suppose you have vectors:

$$

v_1, v_2, v_3
$$

Form matrix A with these vectors as columns.

Then solve:

$$

A\mathbf{c}=0

where

\mathbf{c}=
\begin{bmatrix}
c_1\\
c_2\\
c_3
\end{bmatrix}
$$

This is just another way of writing:

$$

c_1v_1+c_2v_2+c_3v_3=0

$$

Then row reduce A.


What to look for after row reduction

This is the key.

Case 1: every column has a pivot

Then there are no free variables.

So only trivial solution.

So the vectors are independent.

Case 2: at least one column has no pivot

Then there is a free variable.

That means infinitely many solutions.

So there is a non-trivial solution.

So the vectors are dependent.


Why pivots matter

A pivot means that variable is forced.

No pivot means that variable is free.

A free variable lets you choose a non-zero value, which creates a non-trivial solution.

That is exactly dependence.

So:

- all columns pivot → independent
- missing pivot → dependent


Example 6: row reduction step by step

Test whether these are independent:

$$
v_1=
\begin{bmatrix}
1\\2\\1
\end{bmatrix},
\quad
v_2=
\begin{bmatrix}
0\\1\\1
\end{bmatrix},
\quad
v_3=
\begin{bmatrix}
2\\5\\3
\end{bmatrix}
$$

Step 1: form matrix


$$
A=
\begin{bmatrix}
1 & 0 & 2\\
2 & 1 & 5\\
1 & 1 & 3
\end{bmatrix}
$$

Step 2: row reduce

Start:

$$

\begin{bmatrix}
1 & 0 & 2\\
2 & 1 & 5\\
1 & 1 & 3
\end{bmatrix}
$$

Do:

$$

R_2 \leftarrow R_2 - 2R_1 \\

R_3 \leftarrow R_3 - R_1

$$

Result:

$$

\begin{bmatrix}
1 & 0 & 2\\
0 & 1 & 1\\
0 & 1 & 1
\end{bmatrix}

$$

Now:

$$

R_3 \leftarrow R_3 - R_2

$$

$$

\begin{bmatrix}
1 & 0 & 2\\
0 & 1 & 1\\
0 & 0 & 0
\end{bmatrix}
$$

There are only 2 pivots, but 3 columns.

So one variable is free.

Therefore dependent.

Why this example works

Because one of the vectors is a combination of the others.

Let’s check:

From the reduced system, we can expect dependence.

Indeed,

$$

v_3 = 2v_1 + v_2

$$

because

$$
2
\begin{bmatrix}

\end{bmatrix}
+
\begin{bmatrix}

\end{bmatrix}

\begin{bmatrix}

\end{bmatrix}

$$

So the third vector was not new.


### Determinant test

This is a shortcut.

It works only when the matrix is square.

That means:

- 2 vectors in 2D
- 3 vectors in 3D
- n vectors in n-dimensional space

Rule

If $\det(A)\neq 0$, the columns are independent.
If $\det(A)=0$, the columns are dependent.

Why determinant works
A non-zero determinant means the matrix is invertible. If the matrix is invertible, the equation

$$
A\mathbf{c}=0
$$

has only the zero solution. So the columns are independent.
If determinant is zero, matrix is singular, so there are non-zero solutions.
Then the columns are dependent.

Determinant test

Test:

$$
v_1=
\begin{bmatrix}
1\\2
\end{bmatrix},
\quad
v_2=
\begin{bmatrix}
3\\5
\end{bmatrix}
$$

Matrix:

$$

A=
\begin{bmatrix}
1 & 3\\
2 & 5
\end{bmatrix}
$$

Determinant:

$$

\det(A)=1\cdot 5 - 3\cdot 2 = 5-6=-1
$$

Non-zero.

So independent.

Why this works

Because the two vectors do not lie on the same line.

They point in different directions.

Quick tests you should memorize

These are powerful shortcuts.

⸻

1. A single non-zero vector is always independent

Why?

Because

cv=0

can only happen with c=0 if $v\neq 0$. So a non-zero vector by itself is always independent.


2. A set containing the zero vector is always dependent

Why?

Because if one vector is zero, you can write:

$$
0v_1 + 0v_2 + 5(0)=0
$$

This uses a non-zero coefficient.

So you found a non-trivial solution.

Dependent.


3. More vectors than dimension means dependent

In $\mathbb{R}^2$, more than 2 vectors must be dependent.
In $\mathbb{R}^3$, more than 3 vectors must be dependent.

Why?

Because the space does not have enough independent directions.

This is a very important shortcut.

Problem 3: Two vectors in 2D

Test whether

$$

v_1=
\begin{bmatrix}
1\\
2
\end{bmatrix},
\quad
v_2=
\begin{bmatrix}
3\\
4
\end{bmatrix}
$$

are linearly independent.

How to think

For two vectors, first check:

Is one a scalar multiple of the other?

If yes, dependent.
If no, independent.

Here:

* if v_2 = kv_1, then we would need
    3 = k(1), \quad 4 = k(2)
* from the first, k=3
* from the second, k=2

These do not match.

So v_2 is not a scalar multiple of v_1.

Formal test

Solve:

$$
c_1
\begin{bmatrix}
1
2
\end{bmatrix}
+
c_2
\begin{bmatrix}
3
4
\end{bmatrix}
=
\begin{bmatrix}
0
0
\end{bmatrix}
$$

This gives:

$$

\begin{cases}
c_1+3c_2=0\\
2c_1+4c_2=0
\end{cases}

$$

From the first:

$$
c_1=-3c_2
$$

Substitute into the second:

$$
2(-3c_2)+4c_2=0 

-6c_2+4c_2=0

-2c_2=0
\Rightarrow c_2=0

$$

Then:

$$
c_1=0
$$

Only trivial solution.

Result

Linearly independent

Why this works

These two vectors point in different directions. Neither is a stretched copy of the other.

The pattern you should remember forever

Whenever you see a problem, ask these in order:

1. Is there a zero vector?

If yes, dependent.

2. Is there only one non-zero vector?

Then independent.

3. Are there two vectors only?

Check whether one is a scalar multiple of the other.

4. Are there more vectors than the dimension?

Then dependent.

5. Does one vector obviously look like a sum or combination of others?

Then dependent.

6. If nothing is obvious, form a matrix and row reduce.

* every column gets a pivot → independent
* missing pivot / free variable → dependent
