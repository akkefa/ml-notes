# Solution Set of a Linear System

So understanding solution sets helps us understand whether the model parameters are:

- impossible to solve exactly
- uniquely determined
- underdetermined with infinitely many possible solutions

This gives us strong intuition about model training, feature dependence, overfitting, underdetermined systems, and optimization.




For a linear system, there are three possible outcomes:

1. No solution

The equations contradict each other.

2. Unique solution

There is exactly one solution.

3. Infinitely many solutions

There are many possible solutions because some variables are free.

There is no case where a linear system has exactly 2, 3, or 100 solutions.
A linear system has either:

$$
0,\quad 1,\quad \text{or infinitely many solutions}
$$


## Why RREF helps us understand the solution

The original system may look complicated. But after converting it into reduced row echelon form, or RREF, the system becomes much easier to read.

RREF is like the cleanest version of the system.

It helps us answer:

1. Does a solution exist?
2. If a solution exists, is it unique?
3. Which variables are fixed?
4. Which variables are free?


## Pivot variables and free variables

To understand the solution set, we need two terms:

Pivot variable

A pivot variable is a variable whose column contains a pivot.

A pivot is the leading 1 in a nonzero row of the RREF matrix.

For example, if the column for x_1 has a pivot, then x_1 is a pivot variable.

Pivot variables are controlled by equations. Their values are usually determined by other variables or constants.


### Free variable

A free variable is a variable whose column does not contain a pivot.

A free variable can take any value.

This is extremely important.

When a system has free variables, it means the solution is not unique. We can choose different values for the free variables, and each choice gives a valid solution.


### Simple intuition

Think of pivot variables as variables that are “locked” by the equations.

Think of free variables as variables that are “open” or “adjustable.”

If every variable is locked, there is one unique solution.

If at least one variable is free, there are infinitely many solutions.


Case 1: No solution

A system has no solution when the RREF contains a contradiction.

For example, suppose the RREF contains a row like this:

$$
\begin{bmatrix}
0 & 0 & 0 & 1
\end{bmatrix}
$$

If this is an augmented matrix, this row means:

$$
0x_1 + 0x_2 + 0x_3 = 1
$$

which becomes:

0 = 1

This is impossible.

So the system has no solution.

A system with no solution is called an inconsistent system.


ML intuition for no solution

In machine learning, this often happens when the data cannot be perfectly explained by a simple linear model.

For example, suppose we want:

$$
Xw = y
$$

But the data has noise or contradictions. Then there may be no exact weight vector w that satisfies all equations perfectly.

This is common in real-world ML.

That is why in ML we usually solve:

$$
Xw \approx y
$$

instead of requiring:

$$
Xw = y
$$

We use methods such as least squares to find the best approximate solution.

So “no exact solution” does not mean ML fails. It means we need approximation and optimization.


Case 2: Infinitely many solutions

A system has infinitely many solutions when:

1. The system is consistent.
2. There is at least one free variable.

For example, consider this RREF:

$$

\begin{bmatrix}
1 & 0 & 2 & 1 \\
0 & 1 & 1 & 5
\end{bmatrix}
$$

This represents:

$$
x_1 + 2x_3 = 1

x_2 + x_3 = 5
$$

There are three variables:

$$
x_1, x_2, x_3
$$

But only x_1 and x_2 are pivot variables.
The variable x_3 is free.

So we can choose:

x_3 = t

where t can be any real number.

Now solve for the pivot variables.

From:

x_1 + 2x_3 = 1

we get:

x_1 = 1 - 2x_3

Since x_3 = t:

x_1 = 1 - 2t

From:

x_2 + x_3 = 5

we get:

x_2 = 5 - x_3

So:

x_2 = 5 - t

Therefore, the full solution is:

(x_1, x_2, x_3) = (1 - 2t, 5 - t, t)

where:

t \in \mathbb{R}

Because t can be infinitely many values, the system has infinitely many solutions.

ML intuition for infinite solutions

Infinitely many solutions happen when the data does not contain enough independent information to determine all parameters uniquely.

This can happen when:

* there are more features than data points
* some features are linearly dependent
* some columns in the dataset carry repeated information
* the model is over-parameterized

For example, deep neural networks often have millions or billions of parameters. In many cases, there are many different parameter settings that can fit the training data well.

This is related to the idea of overparameterization.

In ML, infinitely many possible solutions are not always bad. But we need a way to choose a good one.

That is why we use:

* regularization
* optimization algorithms
* loss functions
* constraints
* early stopping
* priors
* inductive bias

These help the model choose useful solutions from many possible ones.

⸻

Case 3: Unique solution

A system has a unique solution when:

1. The system is consistent.
2. Every variable is a pivot variable.
3. There are no free variables.

For example:

\begin{bmatrix}
1 & 0 & 0 & 3 \\
0 & 1 & 0 & 5 \\
0 & 0 & 1 & 3
\end{bmatrix}

This represents:

x_1 = 3

x_2 = 5

x_3 = 3

So the solution is:

(x_1, x_2, x_3) = (3, 5, 3)

There is exactly one solution.

⸻

ML intuition for unique solution

A unique solution means the data gives enough independent information to determine the unknown parameters exactly.

In machine learning, this is like saying:

The dataset has enough independent constraints to identify one exact parameter vector.

For example, in linear regression, if X^TX is invertible, the ordinary least squares solution is unique:

w = (X^TX)^{-1}X^Ty

This happens when the feature columns are independent enough.

But in real ML, exact uniqueness is not always required. Many modern models can still work well even when there are many possible parameter solutions.

⸻

Homogeneous linear systems

A homogeneous system has the form:

Ax = 0

Here, the right-hand side is the zero vector.

A homogeneous linear system is always consistent because at least one solution always exists:

x = 0

This is called the trivial solution.

⸻

Trivial solution

The trivial solution is:

x = 0

For example:

A
\begin{bmatrix}
0 \\
0 \\
0
\end{bmatrix}
=
\begin{bmatrix}
0 \\
0 \\
0
\end{bmatrix}

So x=0 always satisfies Ax=0.

⸻

Non-trivial solutions

A homogeneous system may also have non-zero solutions.

These are called non-trivial solutions.

A homogeneous system has non-trivial solutions if there are free variables.

That means:

* if every variable is a pivot variable, only the zero solution exists
* if there is at least one free variable, infinitely many solutions exist

⸻

Why this matters in ML and DL

1. It explains whether model parameters are identifiable

In ML, we often try to learn parameters from data.

If a system has a unique solution, the parameters are identifiable.

If a system has infinitely many solutions, multiple parameter settings can explain the same data.

This is common in deep learning.

⸻

2. It explains underdetermined and overdetermined systems

Underdetermined system

More unknowns than equations.

Example:

\text{more model parameters than data points}

This often creates infinitely many solutions.

Deep learning models are often highly underdetermined.

⸻

Overdetermined system

More equations than unknowns.

Example:

\text{more data points than model parameters}

This often has no exact solution when data is noisy.

In this case, ML finds an approximate solution.

⸻

3. It connects to least squares

Real-world data is noisy, so exact systems like:

Xw = y

may have no exact solution.

Instead, we solve:

\min_w \|Xw - y\|^2

This means:

Find the w that makes the prediction error as small as possible.

This is the foundation of linear regression.

So understanding solution sets helps you understand why least squares exists.

⸻

4. It connects to feature dependence

If one feature is a linear combination of other features, then the dataset contains redundant information.

For example:

x_3 = 2x_1 + 5x_2

Then x_3 does not add truly new information.

This can cause:

* non-unique solutions
* unstable weights
* multicollinearity
* poor interpretability

This is important in regression and feature engineering.


5. It connects to neural networks

Neural networks are trained by solving optimization problems, not usually exact linear systems.

But linear systems still appear inside deep learning concepts:

* linear layers
* matrix multiplication
* gradient updates
* least-squares subproblems
* attention projections
* embeddings
* low-rank approximations
* optimization geometry

A neural network layer does:

z = Wx + b

This is a linear transformation plus bias.

Understanding when linear systems have unique, infinite, or no solutions helps you understand what these layers can and cannot represent.
