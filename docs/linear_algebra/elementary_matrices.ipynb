# Elementary Matrices

Think of an elementary matrix as a matrix that stores one row operation. Instead of saying:

    “Swap row 2 and row 3”

we build a matrix that does that operation automatically when we multiply it with another matrix.

So the big idea is: An elementary matrix is a row operation converted into a matrix.

This matters because in linear algebra, machine learning, and deep learning, many operations are just structured transformations of data. Elementary matrices help us understand how transformations, inverses, decomposition, and solving systems work at the matrix level.

In linear algebra, we often do row operations to solve systems like:

$$
Ax = b
$$

For example:

$$
\begin{bmatrix}
2 & 1 \\
4 & 3
\end{bmatrix}
\begin{bmatrix}
x \\
y
\end{bmatrix}
=
\begin{bmatrix}
5 \\
11
\end{bmatrix}
$$


Normally, we use row operations:

1. Swap rows.
2. Multiply a row by a number.
3. Add one row to another row.

Elementary matrices show that these row operations are not random manual tricks. They are actually matrix multiplications.

That means row operations can be represented algebraically. So instead of thinking:

“I am manually changing rows.”

Think:

    “I am multiplying by special transformation matrices.”

That is powerful because it connects row reduction, inverse matrices, LU decomposition, Gaussian elimination, and numerical linear algebra.

## The Core Idea

Suppose we have a matrix A.

If we left-multiply A by an elementary matrix E, then E performs a row operation on A:

$$
EA = A'
$$

Here:

- E is the elementary matrix.
- A is the original matrix.
- A' is the transformed matrix.

The key rule:

Left multiplication changes rows. If you multiply from the left:

$$
EA
$$

you are applying operations to the rows of A. If you multiply from the right:

$$
AE
$$

you are applying operations to the columns of A.

## Mental Model: Elementary Matrix as a Remote Control

Imagine matrix A is a TV screen.

Elementary matrices are buttons on a remote:

Button	Row Operation
Swap button	Swap two rows
Zoom button	Scale one row
Mix button	Add one row to another

When you press the button, the screen changes.

Similarly:

$$  
EA
$$

means: “Apply the row operation stored inside E to matrix A.”

## Three Types of Elementary Matrices

There are only three fundamental row operations.

### Type 1: Row Swap

A row swap exchanges two rows. For example, to swap row 2 and row 3, start with the identity matrix:

$$
I =
\begin{bmatrix}
1 & 0 & 0 \\
0 & 1 & 0 \\
0 & 0 & 1
\end{bmatrix}
$$

Now swap row 2 and row 3:

$$
P_{23} =
\begin{bmatrix}
1 & 0 & 0 \\
0 & 0 & 1 \\
0 & 1 & 0
\end{bmatrix}
$$

This is called a permutation matrix.

Now suppose:

$$
A =
\begin{bmatrix}
a & b \\
c & d \\
e & f
\end{bmatrix}
$$

Then:

$$
P_{23}A =
\begin{bmatrix}
a & b \\
e & f \\
c & d
\end{bmatrix}
$$

Row 2 and row 3 are swapped.

Why this matters Row swaps are important when the current pivot is zero.


$$
\begin{bmatrix}
0 & 2 \\
3 & 4
\end{bmatrix}
$$

If you want to eliminate using the first pivot, the first pivot is 0, which is bad. So you swap rows.

This is extremely important in numerical computing because we often need stable pivots.

In real ML systems, numerical stability matters when solving linear systems, computing inverses, doing matrix factorization, or optimizing models.

```python
import torch

# row 1 and row 2 are swapped using the elementary matrix

m = torch.tensor([[0, 1], [1, 0]])
A = torch.tensor([[1, 2], [3, 4]])
P = m @ A
print(P)
```

### Type 2: Row Scaling

Row scaling multiplies one row by a nonzero number. For example, to multiply row 2 by 3, start with identity:

$$
I =
\begin{bmatrix}
1 & 0 & 0 \\
0 & 1 & 0 \\
0 & 0 & 1
\end{bmatrix}
$$
Change the second diagonal value to 3:

$$
E =
\begin{bmatrix}
1 & 0 & 0 \\
0 & 3 & 0 \\
0 & 0 & 1
\end{bmatrix}
$$

Now suppose:

$$
A =
\begin{bmatrix}
a & b \\
c & d \\
e & f
\end{bmatrix}
$$

Then:

$$
EA =
\begin{bmatrix}
a & b \\
3c & 3d \\
e & f
\end{bmatrix}
$$

Only row 2 is scaled. Why scaling matters

Scaling is used to make pivots equal to 1.

For example:

$$
\begin{bmatrix}
2 & 4
\end{bmatrix}
$$

If you divide row 1 by 2, you get:

$$
\begin{bmatrix}
1 & 2
\end{bmatrix}
$$

This makes solving systems easier.

In ML, scaling also connects to normalization. Although row scaling in elementary matrices is not exactly the same as feature scaling, the deeper idea is similar:

We transform data into a more useful numerical form.

### Type 3: Row Sum / Row Replacement

This operation adds a multiple of one row to another row.

Example:

$$
r_3 = r_3 + 3r_2
$$

The elementary matrix is:

$$
E =
\begin{bmatrix}
1 & 0 & 0 \\
0 & 1 & 0 \\
0 & 3 & 1
\end{bmatrix}
$$

Notice the 3 is in row 3, column 2.

That location means:

“Add 3 times row 2 into row 3.”

Suppose:

$$
A =
\begin{bmatrix}
a & b \\
c & d \\
e & f
\end{bmatrix}
$$

Then:

$$
EA =
\begin{bmatrix}
a & b \\
c & d \\
e + 3c & f + 3d
\end{bmatrix}
$$

Only row 3 changes.

Why row replacement matters

This is the heart of Gaussian elimination.

When solving a system, we use row replacement to create zeros below pivots.

Example:

$$
A =
\begin{bmatrix}
2 & 1 \\
4 & 3
\end{bmatrix}
$$  

To eliminate the 4 below the 2, we do:

$$
r_2 = r_2 - 2r_1
$$

That gives:

$$
\begin{bmatrix}
2 & 1 \\
0 & 1
\end{bmatrix}
$$

This is the main move behind row echelon form.

## The Most Important Formula

The central formula is: EA = A' This means: Left-multiply by an elementary matrix to apply a row operation.

For multiple row operations:$ E_kE_{k-1}\cdots E_2E_1A = A_{\text{ref}}$

This means we apply many row operations one after another until we get row echelon form.

The order matters. $E_2E_1A$

means:

1. First apply E_1 to A.
2. Then apply E_2 to the result.

Matrix multiplication works from right to left when interpreting transformations.

## Elimination Matrix

Instead of applying many elementary matrices one by one:

$$
E_k(E_{k-1}(\cdots(E_2(E_1A))))
$$  

we can combine them:

$$
E = E_kE_{k-1}\cdots E_2E_1
$$

Then:

$$
EA = A_{\text{ref}}
$$

This combined matrix E is called an elimination matrix.

It stores the whole sequence of row operations.

### Mental model

Suppose each elementary matrix is one instruction:

Instruction 1: swap rows
Instruction 2: subtract 2 times row 1 from row 2
Instruction 3: divide row 2 by 5

The elimination matrix is like the full program:

Run all instructions together.

So: E is not one simple row operation anymore. It is the combined effect of many operations.

### Why Elementary Matrices Are Always Invertible

Every elementary row operation can be undone. That means every elementary matrix has an inverse.

This is very important.

### Row swap inverse

If you swap row 2 and row 3, how do you undo it?

Swap row 2 and row 3 again.

So:

$$
P_{23}^{-1} = P_{23}
$$

A row swap matrix is its own inverse.

### Row scaling inverse
Row scaling inverse

If you multiply a row by 3, undo it by multiplying by $\frac{1}{3}$.

So if:

$$
r_i = 3r_i
$$

then the inverse operation is:

$$
r_i = \frac{1}{3}r_i
$$

Important condition:

You cannot scale by zero. Because if you multiply a row by zero, you destroy information. You cannot recover the original row.

That is why elementary row scaling uses a nonzero scalar.

### Row sum inverse

If you do:

$$
r_i = r_i + \alpha r_j
$$

then the inverse is:

$$
r_i = r_i - \alpha r_j
$$

Example:

$$
r_3 = r_3 + 3r_2
$$

Undo it by:

$$
r_3 = r_3 - 3r_2
$$

So every elementary operation is reversible. That is why elementary matrices are always invertible.

### Connection to Matrix Inverse

Elementary matrices give a deep explanation of matrix inverse.Suppose we reduce matrix A to identity using row operations:

$$
E_kE_{k-1}\cdots E_1A = I
$$

Let:

$$
E = E_kE_{k-1}\cdots E_1
$$

Then:

$$  
EA = I
$$

That means:

$$
E = A^{-1}
$$

So the inverse of A is basically:

The combined row operations that transform A into the identity matrix. This is why the augmented matrix method works:

$$
[A \mid I]
$$

After row reducing:

$$
[I \mid A^{-1}]
$$

You are applying the same elementary row operations to both sides.


### Connection to LU Decomposition

Elementary matrices are also behind LU decomposition. LU decomposition writes a matrix as:

$$
A = LU
$$

Where:

- L is a lower triangular matrix.
- U is an upper triangular matrix.

Gaussian elimination transforms A into an upper triangular matrix U.

So:

$$
EA = U
$$

Then:

$$
A = E^{-1}U
$$

The inverse of the elimination operations becomes the L matrix. So elementary matrices explain why LU decomposition works.

This is important because LU decomposition is used to solve systems efficiently.

Instead of repeatedly solving:

Ax = b

from scratch, we decompose A once:

A = LU

Then solve two easier systems:

Ly = b

and:

Ux = y

This is faster and more stable.


### Why This Matters in Machine Learning

In machine learning, we constantly work with matrices:

- Dataset matrix X

- Covariance matrix
- Embedding matrix

Elementary matrices are not always used directly in daily ML code, but they teach the core idea:

Complex matrix transformations are built from simple operations.

###Example 1: Solving Linear Regression

Linear regression tries to solve:

Xw = y

Usually, the closed-form solution is:

$$
w = (X^TX)^{-1}X^Ty
$$

To compute this, we often need matrix inversion or linear system solving.

Under the hood, libraries like NumPy, PyTorch, TensorFlow, and LAPACK do not usually compute the inverse directly in a naive way. They use decompositions such as:


- LU decomposition
- QR decomposition
- Cholesky decomposition
- SVD

Elementary row operations are the foundation of these methods.

So when you understand elementary matrices, you understand the logic behind:

np.linalg.solve(X, y)
torch.linalg.solve(X, y)
np.linalg.inv(A)
torch.linalg.inv(A)

You are not just calling a black box anymore.

### Example 2: Feature Transformations

Suppose your dataset matrix is:

$$
X =
\begin{bmatrix}
\text{user}_1 \\
\text{user}_2 \\
\text{user}_3
\end{bmatrix}
$$

Each row is one data sample.

A row operation changes the samples.

For example:

- Swapping rows changes sample order.
- Scaling a row changes the magnitude of a sample.
- Adding rows mixes samples.

In normal ML preprocessing, we usually do column operations more often than row operations because columns represent features. But the idea is the same:

Matrices transform data. Elementary matrices make that idea visible in the simplest possible form.


### Example 3: Numerical Stability

When ML libraries solve matrix systems, they often use pivoting.

Pivoting means swapping rows to avoid unstable calculations.

Example:

$$
\begin{bmatrix}
0.000001 & 1 \\
1 & 1
\end{bmatrix}
$$

Using 0.000001 as a pivot can create numerical problems.

So we swap rows:

$$
\begin{bmatrix}
1 & 1 \\
0.000001 & 1
\end{bmatrix}
$$

This row swap is represented by a permutation matrix.

This matters in ML because large models and large datasets can produce matrices with numerical instability. Understanding pivoting helps you understand why some matrix computations fail or become unstable.


### Why This Matters in Deep Learning

In deep learning, we mostly think in terms of tensors, gradients, and neural networks. But the foundation is still linear algebra.

A neural network layer is basically:

$$
y = Wx + b
$$

or for a batch:

$$
Y = XW + b
$$

This is matrix transformation.

Elementary matrices are not what we use to train neural networks directly, but they teach the atomic concept of transformation.


### Deep Learning Connection 1: Layers Are Transformations

Each neural network layer transforms data. For example:

$$
h_1 = W_1x + b_1
$$

$$
h_2 = W_2h_1 + b_2
$$

$$
y = W_3h_2 + b_3
$$

This is similar in spirit to:

$$
E_3E_2E_1A
$$

Each matrix changes the representation.

In elementary matrices:

$$
E_1, E_2, E_3
$$

are simple row operation matrices.

In neural networks:

$$
W_1, W_2, W_3
$$

are learned transformation matrices.

The unforgettable connection:

Elementary matrices are hand-designed transformations. Neural network weight matrices are learned transformations.


### Deep Learning Connection 2: Backpropagation Uses Matrix Transformations

During backpropagation, gradients move backward through layers.

Forward pass:

$$
h = Wx
$$

Backward pass uses:

$$
W^T
$$

So matrix operations are used both forward and backward.

Elementary matrices help you understand that matrix multiplication means applying a transformation. Backpropagation is also a chain of transformations, but applied to gradients.


### Deep Learning Connection 3: Attention in Transformers

In transformers, attention uses matrices:

$$
Q = XW_Q

K = XW_K

V = XW_V
$$

Then attention scores are:

$$
QK^T
$$

And output is:

$$
\text{softmax}(QK^T)V
$$

This is not elementary row operation, but the principle is the same:

Matrices encode operations that transform information.

Elementary matrices are the simplest example of this idea.

Once you understand elementary matrices, you can better understand why multiplying by a matrix can:

- Rotate information.
- Scale information.
- Mix information.
- Project information.
- Reorder information.
- Transform representations.


### Practical Python Example

Let’s take a matrix:

$$
A =
\begin{bmatrix}
2 & 1 \\
4 & 3
\end{bmatrix}
$$

We want to eliminate the 4 below the 2.

Operation:

$$
r_2 = r_2 - 2r_1
$$

The elementary matrix is:

$$
E =
\begin{bmatrix}
1 & 0 \\
-2 & 1
\end{bmatrix}
$$

Now multiply:

$$
EA =
\begin{bmatrix}
1 & 0 \\
-2 & 1
\end{bmatrix}
\begin{bmatrix}
2 & 1 \\
4 & 3
\end{bmatrix}
$$

Result:

$$
EA =
\begin{bmatrix}
2 & 1 \\
0 & 1
\end{bmatrix}
$$

So one matrix multiplication performed the row operation.

Python:

import numpy as np
A = np.array([
    [2, 1],
    [4, 3]
])
E = np.array([
    [1, 0],
    [-2, 1]
])
result = E @ A
print(result)

Output:

[[2 1]
 [0 1]]

This is Gaussian elimination written as matrix multiplication.


### Full Example: Solving a System Using Elementary Matrices

Suppose:

$$
2x + y = 5
$$

$$
4x + 3y = 11
$$

Augmented matrix:

$$
[A \mid b] =
\begin{bmatrix}
2 & 1 & 5 \\
4 & 3 & 11
\end{bmatrix}
$$

Step 1:

$$
r_2 = r_2 - 2r_1
$$

Elementary matrix:

$$
E_1 =
\begin{bmatrix}
1 & 0 \\
-2 & 1
\end{bmatrix}
$$

Apply it:

$$  
E_1[A \mid b]
=
\begin{bmatrix}
2 & 1 & 5 \\
0 & 1 & 1
\end{bmatrix}
$$

Now the second row says:

$$
y = 1
$$

First row:

$$
2x + y = 5
$$

Substitute:

$$
2x + 1 = 5
$$

2x = 4

x = 2

Solution:

$$
x = 2,\quad y = 1
$$

The important idea:

The row operation that helped us solve the system was itself a matrix multiplication.


### Why Identity Matrix Is Used to Create Elementary Matrices

The identity matrix is the “do nothing” matrix:

$$
IA = A
$$

So we start from identity because initially we want no change.

Then we apply one row operation to the identity.

That gives us a matrix that stores exactly that operation.

Why?

Because the identity matrix contains the standard basis rows.

When you modify those basis rows, you are defining how rows of another matrix should be transformed.

Simple analogy:

Identity matrix is a blank command template.
Applying a row operation to it creates a reusable command.


### How to Remember Where to Put the Number in Row Sum Matrix

For operation:

$$
r_i = r_i + \alpha r_j
$$

Put $\alpha$ at position:

$$
(i, j)
$$

Meaning:

Target row = i
Source row = j

Example:

$$
r_3 = r_3 + 3r_2
$$

Target row is 3.

Source row is 2.

So put 3 at row 3, column 2:

$$
\begin{bmatrix}
1 & 0 & 0 \\
0 & 1 & 0 \\
0 & 3 & 1
\end{bmatrix}
$$

Memory trick:

Target row tells you the row position.
Source row tells you the column position.

So:

$$
r_{\text{target}} = r_{\text{target}} + \alpha r_{\text{source}}
$$

means put $\alpha$ at:

$$
(\text{target}, \text{source})
$$

### Difference Between Elementary Matrix and Elimination Matrix

Concept	Meaning
Elementary matrix	Stores one row operation
Elimination matrix	Stores multiple row operations
Identity matrix	Stores no operation
Inverse elementary matrix	Stores the reverse row operation

Example:

E_1

may mean:

Subtract 2 times row 1 from row 2.

E_2

may mean:

Divide row 2 by 3.

E_3

may mean:

Swap row 1 and row 2.

Combined:

E = E_3E_2E_1

means:

Do all three operations in sequence.

Elementary matrices are important because they turn row operations into matrix multiplication.

1. Row swap: exchanges two rows.
2. Row scaling: multiplies a row by a nonzero scalar.
