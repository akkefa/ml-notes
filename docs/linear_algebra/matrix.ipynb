# Matrices

A matrix is a rectangular array of numbers arranged in rows and columns. It is a fundamental mathematical object used in various fields such as linear algebra, physics, computer science, and engineering. Matrices can represent systems of linear equations, transformations, and data structures.

## Types of Matrices
1. **Square Matrix**: A matrix with the same number of rows and columns (n x n).
2. **Rectangular Matrix**: A matrix with a different number of rows and columns (m x n, where m ≠ n).
3. **Diagonal Matrix**: A square matrix where all the elements outside the main diagonal are zero.
4. **Identity Matrix**: A square matrix with ones on the main diagonal and zeros elsewhere.
5. **Zero Matrix**: A matrix where all elements are zero.
6. **Symmetric Matrix**: A square matrix that is equal to its transpose.
7. **Skew-Symmetric Matrix**: A square matrix that is equal to the negative of its transpose.
8. **Orthogonal Matrix**: A square matrix whose rows and columns are orthogonal unit vectors.

## Matrix representation of a linear system

System of linear equations in matrix form helps in machine learning and deep learning because it gives us a compact, scalable, and computable way to represent many relationships at once.

The core idea is this: instead of writing out each equation separately, we can represent the entire system of equations as a single matrix equation. This allows us to use powerful linear algebra techniques to solve the system efficiently, even when dealing with large datasets or complex models. A system of linear equations in matrix form helps because machine learning is fundamentally about learning relationships between many variables at once.

For example, consider a system of linear equations:

$$
2x + 3y = 5
$$

$$
4x - y = 1
$$

We can represent this system in matrix form as:

$$
\begin{bmatrix}
2 & 3 \\
4 & -1
\end{bmatrix}
\begin{bmatrix}
x \\
y
\end{bmatrix}
=
\begin{bmatrix}
5 \\
1
\end{bmatrix}
$$

Here, the matrix on the left contains the coefficients of the variables, the vector in the middle contains the variables, and the vector on the right contains the constants from each equation. This compact representation makes it easier to apply linear algebra techniques to find the solution.

In general, a system of linear equations can be written in matrix form as:

$$
A \mathbf{x} = \mathbf{b}
$$

- $A$ is the matrix of coefficients,
- $\mathbf{x}$ is the vector of variables, and
- $\mathbf{b}$ is the vector of constants.

This form is widely used in machine learning and deep learning for tasks such as linear regression, where we need to solve for the best-fitting parameters.

Why this matters

In machine learning, we almost never deal with just one equation. We deal with:

- many features
- many training examples
- many parameters
- many outputs at the same time

Matrix form lets us handle all of that together. For example, if you have a dataset with many features, a linear model is really just a matrix system. Each row can represent one training example, each column can represent one feature, and the model tries to learn weights that map inputs to outputs.

So matrix form is not just a math shortcut. It is the natural language of machine learning.


Suppose you write a system like:

$$
2w_1 - 4w_2 = 9 \\
3w_1 + 7w_2 = 3
$$

This can be rewritten as:

$$
A =
\begin{bmatrix}
2 & -4 \\
3 & 7
\end{bmatrix},
\quad
x =
\begin{bmatrix}
w_1 \\
w_2
\end{bmatrix},
\quad
b =
\begin{bmatrix}
9 \\
3
\end{bmatrix}
$$

and then:

$$
Ax=b
$$

This is powerful because one matrix now stores the structure of the whole system.

In ML, the same thing happens:

- A often represents data or weights
- x often represents parameters or inputs
- b often represents predictions, labels, or transformed outputs



## Matrix Multiplication

Matrix multiplication is a fundamental operation in linear algebra that combines two matrices to produce a third matrix. The product of two matrices A and B is denoted as AB. For matrix multiplication to be defined, the number of columns in the first matrix (A) must be equal to the number of rows in the second matrix (B).

## Dot Product
The dot product (also known as the scalar product) is an operation that takes two equal-length sequences of numbers (usually vectors) and returns a single number. The dot product of two vectors A and B is calculated as the sum of the products of their corresponding elements.

The simplest way to understand matrix multiplication is through dot product. The dot product of two vectors is simply the sum of products of corresponding elements. Consider the vectors

$$
a = (1, 4, 7) \\
b = (3, 1, 2)
$$

The dot product of 

a and b is calculated as follows:

$$
a \cdot b = (1*3 + 4*1 + 7*2) = 21
$$


A general representation of a dot product is

$$
c = \sum_{k=1}^{n} a_k * b_k
$$
​The same dot product can be used to represent matrix multiplication. Consider the matrices A row and B column:

$$
A = \begin{bmatrix}1 & 4 & 7\end{bmatrix}
B = \begin{bmatrix}3 \\ 1 \\ 2\end{bmatrix}
$$
The product of A and B is calculated as follows:

$$
AB = (1*3 + 4*1 + 7*2) = 21
$$  

 This illustrates how matrix multiplication can be viewed as a series of dot products between rows of the first matrix and columns of the second matrix.


```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import seaborn as sns

# Style
sns.set_theme(style="white")

# Canvas
fig, ax = plt.subplots(figsize=(14, 5), dpi=150)
fig.patch.set_facecolor("#eeeeee")
ax.set_facecolor("#eeeeee")
ax.set_xlim(0, 24)
ax.set_ylim(0, 10)
ax.axis("off")

# -------------------------
# A: Column vector
# -------------------------
xA, yA = 0.5, 0.8
wA, hA = 2.1, 7.0
cell_hA = hA / 4

# Outer box
ax.add_patch(
    patches.Rectangle((xA, yA), wA, hA, linewidth=2.5, edgecolor="black", facecolor="#c9dbef")
)

# Internal lines
for i in range(1, 4):
    ax.plot([xA, xA + wA], [yA + i * cell_hA, yA + i * cell_hA], color="black", lw=2)

# Labels
labels_A = [r'$\mathbf{a_1}$', r'$\mathbf{a_2}$', r'$\vdots$', r'$\mathbf{a_m}$']
for i, lab in enumerate(labels_A):
    yc = yA + hA - (i + 0.5) * cell_hA
    ax.text(xA + wA / 2, yc, lab, ha="center", va="center", fontsize=28)

ax.text(xA + wA / 2, yA + hA + 0.9, "A", ha="center", va="bottom", fontsize=28)

# -------------------------
# Dot operator
# -------------------------
ax.text(3.4, 4.3, r'$\bullet$', fontsize=28, ha='center', va='center')

# -------------------------
# B: Row vector
# -------------------------
xB, yB = 4.3, 3.4
wB, hB = 8.2, 1.8
cell_wB = wB / 4

ax.add_patch(
    patches.Rectangle((xB, yB), wB, hB, linewidth=2.5, edgecolor="black", facecolor="#eddcc3")
)

for i in range(1, 4):
    ax.plot([xB + i * cell_wB, xB + i * cell_wB], [yB, yB + hB], color="black", lw=2)

labels_B = [r'$\mathbf{b_1}$', r'$\mathbf{b_2}$', r'$\cdots$', r'$\mathbf{b_p}$']
for i, lab in enumerate(labels_B):
    xc = xB + (i + 0.5) * cell_wB
    ax.text(xc, yB + hB / 2, lab, ha="center", va="center", fontsize=28)

ax.text(xB + wB / 2, yB + hB + 0.8, "B", ha="center", va="bottom", fontsize=28)

# -------------------------
# Equals sign
# -------------------------
ax.text(13.3, 4.3, "=", fontsize=28, ha="center", va="center")

# -------------------------
# C: Matrix result
# -------------------------
xC, yC = 14.3, 0.8
wC, hC = 9.7, 7.0
colsC, rowsC = 4, 4
cell_wC = wC / colsC
cell_hC = hC / rowsC

ax.add_patch(
    patches.Rectangle((xC, yC), wC, hC, linewidth=2.5, edgecolor="black", facecolor="#dddddd")
)

# Grid lines
for i in range(1, colsC):
    ax.plot([xC + i * cell_wC, xC + i * cell_wC], [yC, yC + hC], color="black", lw=2)
for i in range(1, rowsC):
    ax.plot([xC, xC + wC], [yC + i * cell_hC, yC + i * cell_hC], color="black", lw=2)

# Matrix labels
matrix_labels = [
    [r'$\mathbf{a_1 \cdot b_1}$', r'$\mathbf{a_1 \cdot b_2}$', r'$\cdots$', r'$\mathbf{a_1 \cdot b_p}$'],
    [r'$\mathbf{a_2 \cdot b_1}$', r'$\mathbf{a_2 \cdot b_2}$', r'$\cdots$', r'$\mathbf{a_2 \cdot b_p}$'],
    [r'$\vdots$',               r'$\vdots$',               r'$\ddots$', r'$\vdots$'],
    [r'$\mathbf{a_m \cdot b_1}$', r'$\mathbf{a_m \cdot b_2}$', r'$\cdots$', r'$\mathbf{a_m \cdot b_p}$']
]

for r in range(rowsC):
    for c in range(colsC):
        xc = xC + (c + 0.5) * cell_wC
        yc = yC + hC - (r + 0.5) * cell_hC
        ax.text(xc, yc, matrix_labels[r][c], ha="center", va="center", fontsize=28)

ax.text(xC + wC / 2, yC + hC + 0.9, "C", ha="center", va="bottom", fontsize=28)

plt.tight_layout()
plt.show()
```

Below is an example of matrix multiplication where A and B are (3×3) and (3×2) matrix, respectively. The result is matrix C, which is a (3×2) matrix.

```python


sns.set_theme(style="white")

fig, ax = plt.subplots(figsize=(11, 4.1), dpi=150)
bg = "#eeeeee"
fig.patch.set_facecolor(bg)
ax.set_facecolor(bg)
ax.set_xlim(0, 18)
ax.set_ylim(0, 7)
ax.axis("off")

def draw_grid(ax, x, y, nrows, ncols, cell_w=1.65, cell_h=1.15,
              facecolor="#dbe8f5", edgecolor="black", lw=1.8,
              texts=None, fontsize=16, fontweight="bold"):
    width = ncols * cell_w
    height = nrows * cell_h

    ax.add_patch(
        patches.Rectangle((x, y), width, height,
                          facecolor=facecolor, edgecolor=edgecolor, linewidth=lw)
    )

    for i in range(1, ncols):
        ax.plot([x + i * cell_w, x + i * cell_w], [y, y + height], color=edgecolor, lw=lw)
    for i in range(1, nrows):
        ax.plot([x, x + width], [y + i * cell_h, y + i * cell_h], color=edgecolor, lw=lw)

    if texts:
        for r in range(nrows):
            for c in range(ncols):
                tx = x + (c + 0.5) * cell_w
                ty = y + height - (r + 0.5) * cell_h
                ax.text(tx, ty, str(texts[r][c]), ha="center", va="center",
                        fontsize=fontsize, fontweight=fontweight)

    return width, height, cell_w, cell_h

# A matrix
A = [
    [1, 4, 7],
    [2, 5, 8],
    [3, 6, 9]
]
xA, yA = 0.95, 1.5
wA, hA, cwA, chA = draw_grid(
    ax, xA, yA, 3, 3,
    facecolor="#cfe0f2",
    texts=A
)
ax.text(xA + wA/2, yA + hA + 0.55, "A", ha="center", va="bottom", fontsize=24)

# Red dashed highlight on last row of A
pad = 0.18
ax.add_patch(
    patches.Rectangle(
        (xA + pad, yA + pad),
        wA - 2*pad, chA - 2*pad,
        fill=False, edgecolor="red", linewidth=1.8,
        linestyle=(0, (3.5, 3.5)), alpha=0.8
    )
)

# Dot
ax.text(6.6, 3.2, "•", fontsize=30, ha="center", va="center", fontweight="bold")

# B matrix
B = [
    [3, 1],
    [1, 2],
    [2, 3]
]
xB, yB = 7.25, 1.5
wB, hB, cwB, chB = draw_grid(
    ax, xB, yB, 3, 2,
    facecolor="#f3e0c7",
    texts=B
)
ax.text(xB + wB/2, yB + hB + 0.55, "B", ha="center", va="bottom", fontsize=24)

# Blue dashed highlight on last column of B
ax.add_patch(
    patches.Rectangle(
        (xB + cwB + pad, yB + pad),
        cwB - 2*pad, hB - 2*pad,
        fill=False, edgecolor="dodgerblue", linewidth=1.8,
        linestyle=(0, (3.5, 3.5))
    )
)

# Equals sign
ax.text(11.45, 3.2, "=", fontsize=24, ha="center", va="center", fontweight="bold")

# C matrix
C = [
    [21, 30],
    [27, 36],
    [33, 42]
]
xC, yC = 12.35, 1.5
wC, hC, cwC, chC = draw_grid(
    ax, xC, yC, 3, 2,
    facecolor="#dddddd",
    texts=C
)
ax.text(xC + wC/2, yC + hC + 0.55, "C", ha="center", va="bottom", fontsize=24)

# Green dashed highlight on bottom-right cell of C
ax.add_patch(
    patches.Rectangle(
        (xC + cwC + pad, yC + pad),
        cwC - 2*pad, chC - 2*pad,
        fill=False, edgecolor="green", linewidth=1.8,
        linestyle=(0, (3.5, 3.5))
    )
)

plt.show()
```

## Matrix Operations

### Transpose
The transpose of a matrix is an operation that flips a matrix over its diagonal. This means that the rows of the original matrix become the columns in the transposed matrix, and the columns of the original matrix become the rows in the transposed matrix. The transpose of a matrix A is denoted as A^T.

#### Properties of Transpose
1. (A^T)^T = A
2. (A + B)^T = A^T + B^T
3. (cA)^T = cA^T, where c is a scalar
4. (AB)^T = B^T A^T
5. If A is an m x n matrix, then A^T is an n x m matrix.
6. The transpose of a symmetric matrix is the same as the original matrix, i.e., if A is symmetric, then A^T = A.
7. The transpose of an orthogonal matrix is equal to its inverse, i.e., if A is orthogonal, then A^T = A^-1.

### Inverse
The inverse of a matrix A is a matrix denoted as A^-1 such that when it is multiplied by A, it yields the identity matrix I. In other words, A A^-1 = A^-1 A = I. Not all matrices have inverses; a matrix must be square and have a non-zero determinant to be invertible. The inverse of a matrix is used to solve systems of linear equations, among other applications.
#### Properties of Inverse
1. (A^-1)^-1 = A
2. (AB)^-1 = B^-1 A^-1
3. (A^T)^-1 = (A^-1)^T
4. If A is an invertible matrix, then A^-1 is also invertible, and (A^-1)^-1 = A.
5. The inverse of a product of matrices is the product of the inverses in reverse order, i.e., (AB)^-1 = B^-1 A^-1.
6. If A is an orthogonal matrix, then A^-1 = A^T, meaning the inverse of an orthogonal matrix is equal to its transpose.


### Trace
The trace of a square matrix A is the sum of the elements on the main diagonal of the matrix. It is denoted as tr(A). The trace has several important properties and applications in linear algebra, including in the context of eigenvalues and matrix functions.
#### Properties of Trace
1. tr(A + B) = tr(A) + tr(B)
2. tr(cA) = c tr(A), where c is a scalar
3. tr(AB) = tr(BA)
4. The trace of a matrix is equal to the sum of its eigenvalues, counting multiplicities.
5. The trace of a product of matrices is invariant under cyclic permutations, i.e., tr(ABC) = tr(BCA) = tr(CAB).
6. If A is a diagonal matrix, then the trace of A is simply the sum of its diagonal elements, i.e., if A = diag(a_1, a_2, ..., a_n), then tr(A) = a_1 + a_2 + ... + a_n.
