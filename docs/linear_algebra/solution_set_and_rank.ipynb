# Solution Set and Rank

Given a system of equations, how do we know whether it has no solution one solution or infinitely many solutions? Rank gives us the answer.


## First, What Problem Are We Solving?

A system of linear equations can be written as:

$
Ax=b
$

where:

$$
A = \text{coefficient matrix}
$$

$$
x = \text{unknown vector}
$$

$$
b = \text{output / target vector}
$$

Example:

$$
x_1 + 2x_2 = 4
$$

$$
2x_1 + 5x_2 = 9
$$

This can be written as:

$$
A =
\begin{bmatrix}
1 & 2 \\
2 & 5
\end{bmatrix}

x =
\begin{bmatrix}
x_1 \\
x_2
\end{bmatrix}

b =
\begin{bmatrix}
4 \\
9
\end{bmatrix}

So:

Ax=b

The augmented matrix combines A and b:

[A|b] =
\begin{bmatrix}
1 & 2 & 4 \\
2 & 5 & 9
\end{bmatrix}

## Coefficient Matrix vs Augmented Matrix

This distinction is very important. Coefficient Matrix

The coefficient matrix contains only the coefficients of unknown variables:

$$
A =
\begin{bmatrix}
1 & 2 \\
2 & 5
\end{bmatrix}
$$

It represents the left side of the system.

Augmented Matrix

The augmented matrix includes both coefficients and the output vector b:

$$
[A|b] =
\begin{bmatrix}
1 & 2 & 4 \\
2 & 5 & 9
\end{bmatrix}
$$

It represents the full system.

Rank compares these two matrices:

r(A) and r([A|b])

This comparison tells us whether the system is possible or impossible.

## Main Rule You Must Remember

For a system:

Ax=b

with n unknowns:

Case 1: No Solution

$$
r(A) < r([A|b])
$$

The system is inconsistent.

Case 2: Unique Solution

$$
r(A) = r([A|b]) = n
$$

The system is consistent and has no free variables.

Case 3: Infinite Solutions

$$
r(A) = r([A|b]) < n
$$

The system is consistent but has free variables. This is the complete ranking logic.

## The Memory Trick

Remember this:
A tells us what combinations are possible. [A|b] tells us what output we are asking for.

If adding b increases the rank, it means b brings a new independent direction that A cannot produce.

That means:

$$
Ax=b
$$

is impossible.

So:

$$
r(A) < r([A|b])
$$

means:

The target b is outside the space that A can reach. That is why there is no solution.


## Geometric Meaning

In 2D A linear equation represents a line. Two equations represent two lines.

There are three possibilities:

1. Lines intersect once Unique solution.
2. Lines are parallel No solution.
3. Lines are the same line Infinite solutions.


## Why Rank Detects No Solution

Consider:

$$
x_1 + 2x_2 = 3
$$

$$
2x_1 + 4x_2 = 7
$$

Coefficient matrix:

$$
A =
\begin{bmatrix}
1 & 2 \\
2 & 4
\end{bmatrix}
$$

Augmented matrix:

$$
[A|b] =
\begin{bmatrix}
1 & 2 & 3 \\
2 & 4 & 7
\end{bmatrix}
$$

The rows of A are dependent because:

$$
[2,4] = 2[1,2]
$$

So:

$$
r(A)=1
$$

But in the augmented matrix:

$$
[2,4,7] \neq 2[1,2,3]
$$

because:

$$
2[1,2,3] = [2,4,6]
$$

not:

$$
[2,4,7]
$$

So the augmented matrix has two independent rows.

Therefore:

$$
r([A|b])=2
$$

Now:

$$
r(A) < r([A|b])
$$

So the system has no solution.The augmented matrix reveals the contradiction.


## RREF View of No Solution

The system:

$$
x_1 + 2x_2 = 3
$$

$$
2x_1 + 4x_2 = 7
$$

has augmented matrix:

$$
\begin{bmatrix}
1 & 2 & 3 \\
2 & 4 & 7
\end{bmatrix}
$$

Use row operation:

$$  
R_2 \leftarrow R_2 - 2R_1
$$

$$
\begin{bmatrix}
1 & 2 & 3 \\
0 & 0 & 1
\end{bmatrix}

$$
The second row means:

$$
0x_1 + 0x_2 = 1
$$

which becomes:

0 = 1

Impossible.

So no solution.

This is the strongest sign of inconsistency:

$$
[0 \quad 0 \quad 1]
$$

or generally:

$$
[0 \quad 0 \quad \cdots \quad 0 \mid c]
$$

where:

$$
c \neq 0
$$

This means:

$$
0=c
$$

which is impossible.

```python
import sympy as sp
A = sp.Matrix([
    [1, 2],
    [2, 4]
])
b = sp.Matrix([
    [3],
    [7]
])
augmented = A.row_join(b)
print("Coefficient Matrix A:")
print(A)
print("\nAugmented Matrix [A|b]:")
print(augmented)
print("\nRREF of A:")
print(A.rref()[0])
print("\nRREF of [A|b]:")
print(augmented.rref()[0])
print("\nRank of A:", A.rank())
print("Rank of [A|b]:", augmented.rank())
if A.rank() < augmented.rank():
    print("No solution: system is inconsistent.")
```



9. Unique Solution

A system has a unique solution when:

r(A)=r([A|b])=n

This means two things.

First:

r(A)=r([A|b])

So the system is consistent.

Second:

r(A)=n

So every variable has a pivot.

There are no free variables.

Therefore, there is exactly one solution.

x_1 + 2x_2 = 4

2x_1 + 5x_2 = 9

Coefficient matrix:

A =
\begin{bmatrix}
1 & 2 \\
2 & 5
\end{bmatrix}

Augmented matrix:

[A|b] =
\begin{bmatrix}
1 & 2 & 4 \\
2 & 5 & 9
\end{bmatrix}

Here:

r(A)=2

r([A|b])=2

Number of variables:

n=2

So:

r(A)=r([A|b])=n

Therefore, unique solution.

The solution is:

x_1=2

x_2=1

Check:

2 + 2(1)=4

2(2)+5(1)=9

Both equations are true.

11. RREF View of Unique Solution

Augmented matrix:

\begin{bmatrix}
1 & 2 & 4 \\
2 & 5 & 9
\end{bmatrix}

RREF:

\begin{bmatrix}
1 & 0 & 2 \\
0 & 1 & 1
\end{bmatrix}

This means:

x_1=2

x_2=1

Every variable column has a pivot.

So there are no free variables.

That is why the solution is unique.

13. Infinite Solutions

A system has infinite solutions when:

r(A)=r([A|b])<n

This means:

First:

r(A)=r([A|b])

So the system is consistent.

Second:

r(A)<n

So not every variable has a pivot.

Some variables are free.

Free variables can take infinitely many values.

Therefore, the system has infinite solutions.

⸻

14. Example of Infinite Solutions

x_1 + 2x_2 = 4

2x_1 + 4x_2 = 8

Coefficient matrix:

A =
\begin{bmatrix}
1 & 2 \\
2 & 4
\end{bmatrix}

Augmented matrix:

[A|b] =
\begin{bmatrix}
1 & 2 & 4 \\
2 & 4 & 8
\end{bmatrix}

The second equation is exactly twice the first equation.

So there is really only one independent equation.

Therefore:

r(A)=1

r([A|b])=1

Number of variables:

n=2

So:

r(A)=r([A|b])<n

Therefore, infinite solutions.

⸻

15. RREF View of Infinite Solutions

Augmented matrix:

\begin{bmatrix}
1 & 2 & 4 \\
2 & 4 & 8
\end{bmatrix}

RREF:

\begin{bmatrix}
1 & 2 & 4 \\
0 & 0 & 0
\end{bmatrix}

This means:

x_1 + 2x_2 = 4

There is a pivot in the x_1 column, but no pivot in the x_2 column.

So x_2 is free.

Let:

x_2=s

Then:

x_1 + 2s = 4

x_1 = 4 - 2s

So the solution set is:

x_1 = 4 - 2s

x_2 = s

or:

x =
\begin{bmatrix}
4 - 2s \\
s
\end{bmatrix}

We can split it:

x =
\begin{bmatrix}
4 \\
0
\end{bmatrix}
+
s
\begin{bmatrix}
-2 \\
1
\end{bmatrix}

This means the solution set is a line.

⸻

17. Homogeneous Systems

A homogeneous system has this form:

Ax=0

That means the right side is the zero vector.

Example:

x_1 + 2x_2 = 0

2x_1 + 5x_2 = 0

The important point:

A homogeneous system is always consistent.

Why?

Because:

x=0

always works.

If:

x =
\begin{bmatrix}
0 \\
0 \\
\vdots \\
0
\end{bmatrix}

then:

Ax=0

So every homogeneous system has at least one solution.

This is called the trivial solution.

⸻

18. Trivial Solution

The trivial solution is:

x=0

For example:

x_1 = 0

x_2 = 0

x_3 = 0

This solution always satisfies:

Ax=0

because multiplying any matrix by the zero vector gives the zero vector.

⸻

19. Homogeneous System: Unique Trivial Solution

A homogeneous system has only the trivial solution when:

r(A)=n

Since:

r([A|0])=r(A)

the condition can be written as:

r([A|0])=r(A)=n

This means every variable has a pivot.

There are no free variables.

So only:

x=0

is possible.

⸻

x_1 + 2x_2 = 0

2x_1 + 5x_2 = 0

Coefficient matrix:

A =
\begin{bmatrix}
1 & 2 \\
2 & 5
\end{bmatrix}

Here:

r(A)=2

Number of variables:

n=2

So:

r(A)=n

Therefore, only the trivial solution exists.

Solution:

x_1=0

x_2=0

21. Homogeneous System: Infinite Solutions

A homogeneous system has infinite solutions when:

r(A)<n

Because:

r(A)=r([A|0])

the system is always consistent.

If rank is less than the number of variables, then there are free variables.

Free variables create infinitely many solutions.

⸻

24. Why Homogeneous Systems Are Special

For nonhomogeneous systems:

Ax=b

we have three possibilities:

1. No solution
2. Unique solution
3. Infinite solutions

But for homogeneous systems:

Ax=0

we only have two possibilities:

1. Only trivial solution
2. Infinite solutions

There is never “no solution” because x=0 always works.

26. Deep Intuition: Rank Measures Constraints

Each independent equation gives one independent constraint.

If you have n unknowns and rank r, then roughly:

\text{degrees of freedom} = n-r

For consistent systems:

\text{number of free variables} = n-r(A)

So:

* if n-r=0, unique solution
* if n-r>0, infinite solutions

Example:

If there are 5 unknowns and rank is 5:

5-5=0

No free variables.

Unique solution.

If there are 5 unknowns and rank is 3:

5-3=2

Two free variables.

Infinite solutions.

This is a very important idea.

⸻

27. Connection to Null Space

For homogeneous systems:

Ax=0

the solution set is called the null space of A.

N(A)=\{x: Ax=0\}

The dimension of the null space is:

\text{nullity}(A)=n-r(A)

This is called the rank-nullity theorem:

\text{rank}(A)+\text{nullity}(A)=n

Meaning:

\text{independent constraints} + \text{free directions} = \text{total variables}

This is one of the most important theorems in linear algebra.

⸻

28. Example: Rank-Nullity

Suppose:

A \in \mathbb{R}^{3 \times 5}

and:

r(A)=3

Then:

\text{nullity}(A)=5-3=2

So the homogeneous system:

Ax=0

has a 2-dimensional solution space.

That means there are infinitely many solutions controlled by two free parameters.

29. Why This Matters in Machine Learning

This concept is everywhere in ML.

When we train a model, we often solve equations approximately:

Xw=y

where:

* X is the feature matrix
* w is the model parameters
* y is the target

Rank tells us whether the data has enough independent information to determine the parameters uniquely.

⸻

30. ML Example: Linear Regression

In linear regression:

Xw=y

If X has full column rank, the model parameters may be uniquely determined.

If X does not have full column rank, some features are redundant.

Then many different w values can give the same predictions.

Example:

Suppose:

x_2 = 2x_1

Then:

w_1x_1 + w_2x_2

becomes:

w_1x_1 + 2w_2x_1

= (w_1 + 2w_2)x_1

Many pairs (w_1,w_2) can produce the same result.

So the solution is not unique.

This is a rank problem.

⸻

31. ML Meaning of No Solution

In real ML data, often:

Xw=y

has no exact solution.

Why?

Because data is noisy.

For example, two training samples may have the same features but different targets:

Feature x	Target y
10	100
10	120

The model cannot satisfy both exactly with a deterministic linear function.

So the system is inconsistent.

That means:

r(X)<r([X|y])

In ML, instead of demanding an exact solution, we use least squares:

\min_w ||Xw-y||^2

So we find the best approximate solution.

This is one of the main reasons optimization exists in ML.

⸻

32. ML Meaning of Infinite Solutions

Modern ML models often have more parameters than data points.

That means:

n > m

More unknowns than equations.

This usually gives infinite solutions if the system is consistent.

In deep learning, this is common.

The model may have millions or billions of parameters and still fit the training data perfectly.

There may be many parameter settings that produce the same training predictions.

This is connected to:

* overparameterization
* flat minima
* generalization
* implicit bias of gradient descent
* regularization
* weight decay
* minimum norm solutions

33. Why Rank Matters in Deep Learning

Neural networks contain many matrix multiplications:

h = Wx+b

The rank of W tells us how much information can pass through that layer.

If W has low rank, it creates a bottleneck.

If the activations or embeddings collapse into low rank, the network may lose expressive power.

Rank helps us understand:

* representation collapse
* bottleneck layers
* LoRA
* adapters
* compression
* overparameterized models
* linear dependency in features
* whether embeddings are diverse

⸻

34. Deep Learning Example: Bottleneck Layer

Suppose a network has:

1000 \rightarrow 10 \rightarrow 1000

The middle layer has dimension 10.

Even though the output returns to 1000 dimensions, the transformation can pass through at most 10 independent directions.

So the effective rank is at most 10.

This means the network is forced to compress information.

This idea is used in:

* autoencoders
* representation learning
* bottleneck architectures
* dimensionality reduction

⸻

35. Deep Learning Example: LoRA

LoRA updates a large weight matrix using a low-rank update:

W' = W + BA

where:

B \in \mathbb{R}^{d \times r}

A \in \mathbb{R}^{r \times k}

The update:

BA

has rank at most r.

So if:

r=8

then the update only changes the model in 8 independent directions.

This makes fine-tuning much cheaper.

Rank explains why LoRA works.

36. Important Connection to Recommendation Systems

In recommendation systems, we often have a user-item interaction matrix:

R \in \mathbb{R}^{m \times n}

where:

* rows are users
* columns are items
* entries are clicks, ratings, purchases, or views

Usually this matrix is huge but low-rank.

Why?

Because user behavior is controlled by hidden factors.

For example:

* price preference
* location preference
* category preference
* brand preference
* freshness preference
* popularity preference
* intent
* urgency

So we approximate:

R \approx UV^T

where:

U = \text{user embeddings}

V = \text{item embeddings}

This is matrix factorization.

Rank controls how many hidden dimensions we use.

If the rank is too small, the model underfits.

If the rank is too large, the model may overfit.

x_1, x_2, x_3

Each independent equation reduces your freedom by one dimension.

So:

3 variables, rank 3

3-3=0

No freedom left.

One point.

Unique solution.

One freedom left.

A line of solutions.

Infinite solutions.

3-1=2

Infinite solutions.

Contradiction

If the equations contradict each other, there is no solution.

That happens when:

38. Full Python Function to Classify Any Linear System

import sympy as sp
def classify_linear_system(A_values, b_values):
    A = sp.Matrix(A_values)
    b = sp.Matrix(b_values)
    
    if b.shape[1] != 1:
        b = b.reshape(b.rows * b.cols, 1)
    
    augmented = A.row_join(b)
    
    rank_A = A.rank()
    rank_aug = augmented.rank()
    n = A.shape[1]
    
    print("Coefficient matrix A:")
    print(A)
    
    print("\nVector b:")
    print(b)
    
    print("\nAugmented matrix [A|b]:")
    print(augmented)
    
    print("\nRREF of [A|b]:")
    print(augmented.rref()[0])
    
    print("\nRank of A:", rank_A)
    print("Rank of [A|b]:", rank_aug)
    print("Number of variables:", n)
    
    if rank_A < rank_aug:
        return "No solution"
    elif rank_A == rank_aug == n:
        return "Unique solution"
    else:
        return "Infinite solutions"

)

40. Final Memory Line

The whole topic can be remembered like this:

Rank tells us how many independent constraints we have.
Variables tell us how many freedoms we started with.
If constraints contradict the target, there is no solution.
If constraints fully control all variables, there is one solution.
If some variables remain free, there are infinitely many solutions.

Mathematically:

\boxed{\text{free variables} = n - r(A)}

for every consistent system.

And that is why rank is the key to understanding solution sets.
