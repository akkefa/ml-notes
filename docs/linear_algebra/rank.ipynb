# Rank

The rank of a matrix tells us how much real independent information exists inside that matrix.

A matrix may look large, but many of its rows or columns may be repeated, redundant, or created from combinations of others.

So rank answers this question:

:::{note}
“How many genuinely independent directions or pieces of information does this matrix contain?”
:::

For a matrix A, the rank is written as: r(A)

$$
\text{rank}(A) = \text{number of linearly independent rows or columns}
$$

Even though we can define row rank and column rank separately, an important theorem says:

$$
\text{row rank} = \text{column rank}
$$

So we simply say rank.

## Simple Intuition: Rank Means “True Dimension”

Imagine you have the following three vectors:

$$
v_1 = [1, 2]
$$

$$
v_2 = [2, 4]
$$

$$
v_3 = [3, 6]
$$

At first, it looks like we have three vectors. But actually:

$$
v_2 = 2v_1
$$

$$
v_3 = 3v_1
$$

So all three vectors are pointing in the same direction. They do not give us three independent directions. They only give us one direction.

So the rank is:

1

That means the data has only one true dimension.

This is the heart of rank.


## A Memorable Analogy

Think of a team meeting. There are 10 people in the room. But if 7 people only repeat what the first 3 people already said, then the meeting has only 3 independent ideas.

So:

$$
\text{number of people} = 10
$$

but

$$
\text{rank of ideas} = 3
$$

Matrix rank works the same way. A matrix may have many rows and columns, but rank tells us how many independent “ideas” or “directions” actually exist.

## Why Rank Matters in Linear Systems

Suppose we have a system:

Ax = b

Here:

- A is the coefficient matrix.
- x is the unknown vector.
- b is the output vector.

The rank of A tells us whether the equations are independent or redundant.

For example:

$$
x + y = 2
$$

$$
2x + 2y = 4
$$

The second equation is just twice the first equation. So this system does not contain two independent equations. It contains only one independent equation. The rank of the coefficient matrix is 1.

That means we do not have enough independent information to get a unique solution.

So rank helps us answer:

1. Does the system have a solution?
2. Is the solution unique?
3. Are some equations redundant?
4. Are some variables free?

## Rank and RREF

The rank of a matrix equals the number of pivots in its reduced row echelon form.

$$
r(A) = \text{number of pivots in RREF of } A
$$

A pivot represents an independent row or column. When we reduce a matrix to RREF, dependent rows become zero rows.

$$
A =
\begin{bmatrix}
1 & 2 \\
2 & 4 \\
3 & 6
\end{bmatrix}
$$

RREF becomes:

$$
\begin{bmatrix}
1 & 2 \\
0 & 0 \\
0 & 0
\end{bmatrix}
$$

There is only one pivot.

So:

r(A) = 1

This means only one row is truly independent.

```python
import torch

A = torch.tensor([
    [1, 2],
    [2, 4],
    [3, 6]
], dtype=torch.float32)
rank = torch.linalg.matrix_rank(A)
print("Matrix A:")
print(A)
print("Rank of A:", rank)
```

## Full Row Rank
For a matrix:

$$
A \in \mathbb{R}^{m \times n}
$$

it has full row rank if:

$$
r(A) = m
$$

This means all rows are linearly independent.

$$
A =
\begin{bmatrix}
1 & 0 & 2 \\
0 & 1 & 3
\end{bmatrix}
$$

This is a $2 \times 3$ matrix. It has 2 rows. If both rows are independent, then:

$$
r(A) = 2
$$

So it has full row rank. Important point: A matrix can only have full row rank if:

$$
m \leq n
$$

because rank cannot be larger than the smaller dimension.

$$
r(A) \leq \min(m,n)
$$

## Full Column Rank

A matrix has full column rank if:

$$
r(A) = n
$$

This means all columns are linearly independent.

$$
A =
\begin{bmatrix}
1 & 0 \\
0 & 1 \\
2 & 3
\end{bmatrix}
$$

This is a $3 \times 2$ matrix. it has 2 columns. If both columns are independent:

r(A) = 2

So the matrix has full column rank.

A matrix can only have full column rank if:

$$
n \leq m
$$

Again:

$$
r(A) \leq \min(m,n)
$$

## Full Rank

A matrix is called full rank if its rank is as large as possible.

For an $m \times n$ matrix:

$$
r(A) = \min(m,n)
$$

If:

$$
A \in \mathbb{R}^{5 \times 3}
$$

then maximum possible rank is:

3

So if:

r(A) = 3

then A is full rank.

## Rank and Transpose

The rank of a matrix and its transpose are the same:

$$
r(A) = r(A^T)
$$

Why?

Because rank measures independent information. When you transpose a matrix, rows become columns and columns become rows. But the amount of independent information does not change.

Example:

$$
A =
\begin{bmatrix}
1 & 2 \\
2 & 4 \\
3 & 6
\end{bmatrix}
$$

$$
A^T =
\begin{bmatrix}
1 & 2 & 3 \\
2 & 4 & 6
\end{bmatrix}
$$

Both have rank 1.



```python
A = torch.tensor([
    [1, 2],
    [2, 4],
    [3, 6]
], dtype=torch.float32)
print("Rank of A:", torch.linalg.matrix_rank(A))
print("Rank of A.T:", torch.linalg.matrix_rank(A.T))
```

## Rank and Matrix Dimensions

For any matrix:

A $\in \mathbb{R}^{m \times n}$ rank cannot be greater than the number of rows:

$$
r(A) \leq m
$$

and rank cannot be greater than the number of columns:

$$
r(A) \leq n
$$


$$
r(A) \leq \min(m,n)
$$

If A is a 100 $\times 5$ matrix, maximum rank is 5. Why?

Because there are only 5 columns. You cannot have more than 5 independent column directions. If A is a 5 $\times 100$ matrix, maximum rank is 5. Why? Because there are only 5 rows. You cannot have more than 5 independent row directions.

## Rank of Matrix Product

For two matrices A and B, the rank of their product satisfies:

$$
r(AB) \leq \min(r(A), r(B))
$$

This is extremely important.

It means multiplying matrices cannot create more independent information than what already exists in the original matrices. Matrix multiplication can transform, compress, rotate, or mix information. But it cannot magically create new independent information.

### Row View

In the product:

AB

the rows of AB are linear combinations of the rows of B.

Therefore:

$$
r(AB) \leq r(B)
$$

The product cannot have more independent rows than B.

### Column View

The columns of AB are linear combinations of the columns of A.

Therefore:

$$
r(AB) \leq r(A)
$$

The product cannot have more independent columns than A.

So together:

$$
r(AB) \leq \min(r(A), r(B))
$$


```python
import torch

A = torch.randn(4, 3)
B = torch.randn(3, 5)
AB = A @ B
rank_A = torch.linalg.matrix_rank(A)
rank_B = torch.linalg.matrix_rank(B)
rank_AB = torch.linalg.matrix_rank(AB)
print("Rank of A:", rank_A)
print("Rank of B:", rank_B)
print("Rank of AB:", rank_AB)
print("Minimum rank of A and B:", min(rank_A, rank_B))
print("Is rank(AB) <= min(rank(A), rank(B))?")
print(rank_AB <= min(rank_A, rank_B))
```

## Why Rank Matters in Machine Learning

Rank is not just a math concept. It appears everywhere in machine learning.

It helps us understand:

* feature redundancy
* model capacity
* dimensionality reduction
* embeddings
* neural network layers
* overparameterization
* PCA
* matrix factorization
* recommender systems
* optimization problems
* singular matrices
* generalization

### Rank and Feature Redundancy

Suppose your dataset has features:

House Size	Size in Square Feet	Double Size
Small	1000	2000
Medium	1500	3000
Large	2000	4000

Here, “Double Size” is just:

$2 \times \text{Size in Square Feet}$

So it does not add new information.

If your feature matrix contains many columns like this, then your feature matrix is low-rank. That means your model is receiving redundant features.

In machine learning, this can cause:

* unstable coefficients
* poor interpretability
* multicollinearity
* overfitting
* unnecessary computation

For linear regression, redundant features can make the matrix X^TX singular or nearly singular.

The closed-form solution for linear regression is:

$$
w = (X^TX)^{-1}X^Ty
$$

But if X^TX is not full rank, then inverse may not exist. So rank directly affects whether the linear regression solution is stable.

### Rank and Linear Regression

In linear regression:

y = Xw

where:

- X is the feature matrix
- w is the weight vector
- y is the target

For a unique least-squares solution, X should have full column rank. If X has dependent columns, then multiple weight vectors can produce the same prediction.

Suppose:

$$  
x_2 = 2x_1
$$

Then:

$$
w_1x_1 + w_2x_2
$$

becomes:

$$
w_1x_1 + w_2(2x_1) = (w_1 + 2w_2)x_1
$$

Many combinations of w_1 and w_2 can produce the same result. So the model cannot uniquely decide which feature is responsible.

This is why rank matters.


```python
import numpy as np
X = np.array([
    [1, 2],
    [2, 4],
    [3, 6],
    [4, 8]
])
print("X:")
print(X)
print("Rank of X:", np.linalg.matrix_rank(X))
print("Number of columns:", X.shape[1])
```

### Rank and PCA

PCA finds the main independent directions of variation in data. If your data matrix is low-rank, then most of its information lies in a smaller number of directions.

For example, suppose you have images of faces. Each image may have thousands of pixels.

But the true variation may be controlled by fewer factors:

- lighting
- face angle
- expression
- skin tone
- background
- camera distance

So even though the raw dimension is huge, the data may approximately live in a lower-dimensional space. This is the idea behind PCA.

Rank tells us the maximum number of meaningful principal components.

If:

$$
X \in \mathbb{R}^{1000 \times 500}
$$


$$
r(X) \leq 500
$$

So PCA can produce at most 500 nonzero principal components. If the actual rank is 50, then only 50 components carry independent information.

### Rank and SVD

Singular Value Decomposition decomposes a matrix as:

$$
A = U\Sigma V^T
$$

The rank of A equals the number of nonzero singular values in \Sigma.

If singular values are:

[10, 5, 1, 0, 0]

then:

r(A) = 3

In real ML systems, singular values are often not exactly zero but very small.

Example:

[10, 5, 1, 0.001, 0.00001]

This means the matrix is approximately low-rank.

This idea is used in:

- compression
- denoising
- PCA
- recommender systems
- low-rank adaptation
- matrix factorization


### Rank and Recommender Systems

This is very important for recommendation engines. Suppose we have a user-item rating matrix:

$$
R =
\begin{bmatrix}
5 & 5 & 0 & 0 \\
4 & 4 & 0 & 0 \\
0 & 0 & 5 & 5 \\
0 & 0 & 4 & 4
\end{bmatrix}
$$

Rows are users. Columns are items.

This matrix may look like it has many users and items, but the hidden structure may be simple.

Maybe users belong to only two interest groups:

1. People who like action movies.
2. People who like romance movies.

So the matrix can be approximated using a low-rank factorization:

$$
R \approx UV^T
$$

where:

- U represents user embeddings
- V represents item embeddings

This is the foundation of matrix factorization in recommender systems. Rank controls how many hidden factors we use.

For example:

$$
R \approx U_{m \times k} V_{n \times k}^T
$$

Here k is the embedding dimension.

If k = 10, we are saying:

“User preferences can be explained using 10 hidden dimensions.”

These hidden dimensions may represent things like:

- price sensitivity
- location preference
- category interest
- brand preference
- freshness preference
- popularity preference
- visual style
- urgency
- intent level
- novelty preference

So rank is deeply connected to recommender embeddings.


### Rank and Embeddings

An embedding matrix is usually:

$$  
E \in \mathbb{R}^{N \times d}
$$

where:

- N = number of users/items/tokens/words
- d = embedding dimension

For example:

$$
E \in \mathbb{R}^{1,000,000 \times 768}
$$

The maximum rank is: 768

because:

$$
r(E) \leq \min(1,000,000, 768) = 768
$$

This means even if we have one million items, their embeddings can only span at most 768 independent directions.

That is not bad. That is the whole point of embeddings. Embeddings compress many objects into a smaller semantic space.

Rank helps us understand how much independent variation exists inside that embedding space. If the effective rank is very low, maybe the embeddings are collapsed.

If the rank is high and meaningful, the embeddings capture richer variation.

### Rank and Deep Learning Layers

A neural network linear layer is:

y = Wx + b

where:

$$
W \in \mathbb{R}^{d_{out} \times d_{in}}
$$

The rank of W tells us how much information the layer can preserve or transform. If W is full rank, it can preserve many independent directions.

If W is low-rank, it compresses the input into fewer independent directions.

For example:

$$
W \in \mathbb{R}^{100 \times 100}
$$

If:

$$
r(W) = 100
$$

the layer can represent transformations across 100 independent directions.

But if:

r(W) = 10

then the layer only uses 10 independent directions.This means the layer is acting like a bottleneck.


### Bottleneck Layers and Rank

Many deep learning architectures intentionally use bottlenecks.

Example:

$$
768 \rightarrow 128 \rightarrow 768
$$

The middle dimension is 128. The transformation cannot have rank greater than 128.

Why?

Because:

$$
W_1 \in \mathbb{R}^{128 \times 768}
$$

$$
W_2 \in \mathbb{R}^{768 \times 128}
$$

The product: W_2W_1 has rank at most: 128

So this bottleneck forces the model to compress information.

This is used in:

* autoencoders
* adapters
* LoRA
* projection layers
* dimensionality reduction
* model compression


### Rank and LoRA

LoRA means Low-Rank Adaptation. Instead of updating a large weight matrix W, LoRA freezes W and learns a low-rank update:

$$
W' = W + BA
$$

where:

$$
A \in \mathbb{R}^{r \times d}
$$

$$
B \in \mathbb{R}^{k \times r}
$$

The update matrix: BA has rank at most r. So if r = 8, the model only learns an update in 8 independent directions. This is why LoRA is memory-efficient.

It says:

    “We do not need to change the full matrix. We only need to adjust a small number of important directions.”

That is rank in action.

### Rank and Attention in Transformers

In transformers, attention uses matrices:

$$
Q = XW_Q
$$

$$
K = XW_K
$$

$$
V = XW_V
$$

The rank of these projections affects how much information can be represented in query, key, and value spaces. If projection matrices are low-rank or become rank-deficient, the attention mechanism may lose expressive power.

Also, attention matrices can often have approximately low-rank structure, which is why many efficient transformer methods try to approximate attention using low-rank methods.

### Rank and Overparameterization

Deep networks often have millions or billions of parameters. But many learned transformations are effectively low-rank.
That means the model has many parameters, but the useful learned information may live in a smaller subspace.

This helps explain why:

- pruning can work
- quantization can work
- LoRA can work
- adapters can work
- model compression can work
- some networks generalize despite being huge

Rank helps us analyze the real capacity being used by the model.


### Rank and Data Quality

Rank also helps detect bad data. Suppose you have a dataset with 100 features. But rank is only 20. This means 80 features may be redundant combinations of others.

Possible reasons:

- duplicate columns
- derived features
- one-hot encoding mistakes
- leakage
- constant columns
- highly correlated sensors
- repeated measurements
- bad preprocessing

So checking rank can reveal hidden problems in your feature matrix.

### Rank and One-Hot Encoding

Suppose you one-hot encode a categorical variable with 3 categories:

Color	Red	Green	Blue
Red	1	0	0
Green	0	1	0
Blue	0	0	1

Now if you also add an intercept column of ones:

$$
\text{Intercept} = Red + Green + Blue
$$

So one column becomes dependent on the others. This is called the dummy variable trap. The design matrix becomes rank-deficient. That can cause problems in linear regression.

Solution:

Drop one category.

For example, use only:

- Red
- Green

Then Blue becomes the baseline category.


### Rank and Invertibility

For a square matrix:

$$
A \in \mathbb{R}^{n \times n}
$$

the matrix is invertible only if:

$$
r(A) = n
$$

That means a square matrix is invertible only when it has full rank. If rank is less than n, the matrix is singular and cannot be inverted.

This matters in ML because inverse appears in:

- linear regression
- least squares
- covariance matrices
- Gaussian distributions
- Kalman filters
- optimization
- Newton’s method
- second-order methods

If the matrix is not full rank, the inverse does not exist.

### Rank and Covariance Matrix

A covariance matrix tells how features vary together. If the covariance matrix is low-rank, some features are linearly dependent or nearly dependent.

This matters in:

- PCA
- Gaussian models
- anomaly detection
- Mahalanobis distance
- multivariate statistics

For Mahalanobis distance:

$$
d(x,\mu) = \sqrt{(x-\mu)^T \Sigma^{-1}(x-\mu)}
$$

We need the inverse of covariance matrix $\Sigma$.

$$
If \Sigma is rank-deficient, \Sigma^{-1} does not exist.
$$

So we may need regularization:

$$
\Sigma_{\text{regularized}} = \Sigma + \lambda I
$$

This makes the matrix more stable.

### Rank and Gradient Descent

In optimization, rank affects the curvature of the loss surface.

For linear regression, the Hessian is:

$$
H = X^TX
$$

If X is rank-deficient, then X^TX is also rank-deficient. This means some directions in parameter space are flat.
So gradient descent may have infinitely many solutions. This is common in overparameterized models.
The model can fit the data using many different parameter combinations.

### Rank and Neural Network Expressiveness

Consider two linear layers without activation:

$$
y = W_2W_1x
$$

This is still just one linear transformation.

And:

$$
r(W_2W_1) \leq \min(r(W_1), r(W_2))
$$

So stacking linear layers without nonlinear activation does not create more expressive power. This is one reason nonlinear activations are essential. Without ReLU, GELU, sigmoid, or tanh, a deep network would collapse into one linear layer. Rank helps explain this.

### Important Example: Two Linear Layers

Suppose:

$$  
W_1 \in \mathbb{R}^{10 \times 100}
$$

$$
W_2 \in \mathbb{R}^{100 \times 10}
$$

Then:

$$  
W_2W_1 \in \mathbb{R}^{100 \times 100}
$$

But:

$$
r(W_2W_1) \leq 10
$$

So even though the final matrix is 100 \times 100, its rank is at most 10.

This is a bottleneck.The network compressed the input into 10 dimensions and then expanded it back.But once information is compressed into 10 independent directions, the model cannot recover the lost directions.

This is why bottlenecks must be designed carefully.


```python
import numpy as np
np.random.seed(0)
W1 = np.random.randn(10, 100)
W2 = np.random.randn(100, 10)
W = W2 @ W1
print("Shape of W:", W.shape)
print("Rank of W1:", np.linalg.matrix_rank(W1))
print("Rank of W2:", np.linalg.matrix_rank(W2))
print("Rank of W2 @ W1:", np.linalg.matrix_rank(W))
```

### Rank in NLP

In NLP, we often use word embedding matrices.

Example:

$$
E \in \mathbb{R}^{V \times d}
$$

where:

- V = vocabulary size
- d = embedding dimension

If vocabulary size is 50,000 and embedding dimension is 768:

$$
E \in \mathbb{R}^{50000 \times 768}
$$

The maximum rank is 768. This means all word meanings are represented inside a 768-dimensional semantic space.

Rank helps us understand the expressive capacity of that embedding space.

### Rank and Representation Collapse

In self-supervised learning and contrastive learning, representation collapse is a major problem. Representation collapse means many inputs map to very similar or identical embeddings. If all embeddings become the same vector, the embedding matrix has rank 1.

That is terrible.

A good embedding model should spread examples across meaningful independent directions. So rank can be used as a diagnostic. Low rank may indicate collapsed representations.

High effective rank may indicate more diverse representations.

### Practical ML Meaning

Rank tells us:

Situation	Rank Meaning
Feature matrix rank is low	Features are redundant
Weight matrix rank is low	Layer is compressing information
Embedding matrix rank is low	Embeddings may lack diversity
Covariance matrix rank is low	Features are dependent
User-item matrix is low-rank	Preferences can be explained by hidden factors
Product AB has low rank	Transformation is bottlenecked
Square matrix is full rank	Matrix is invertible
Design matrix is full column rank	Linear regression solution is unique

### Rank vs Shape

A common mistake is thinking matrix size equals information. Not true. A matrix can be huge but low-rank.

Example:

$$
A \in \mathbb{R}^{1000 \times 1000}
$$

But if:

r(A) = 5

then this huge matrix only contains 5 independent directions. So shape tells us storage size.Rank tells us information size.

That is a powerful distinction.

### Rank and Data Compression

If a matrix is low-rank, we can compress it.

Instead of storing:

$$
A \in \mathbb{R}^{m \times n}
$$

we store:

$$
A \approx UV^T
$$

where:

U \in \mathbb{R}^{m \times k}

V \in \mathbb{R}^{n \times k}

If k is much smaller than m and n, storage is greatly reduced.

Original storage:

mn

Compressed storage:

mk + nk

Example:

If:

m = 10000, n = 10000

Original storage:

10000 \times 10000 = 100,000,000

If k = 100:

10000 \times 100 + 10000 \times 100 = 2,000,000

Huge reduction.

This is why low-rank approximation matters.



```python
import torch

# Low-rank matrix
A = torch.tensor([
    [1, 2, 3],
    [2, 4, 6],
    [3, 6, 9]
], dtype=torch.float32)

# Full-rank matrix
B = torch.tensor([
    [1, 2, 3],
    [0, 1, 4],
    [5, 6, 0]
], dtype=torch.float32)
print("Matrix A:")
print(A)
print("Rank of A:", torch.linalg.matrix_rank(A))
print()
print("Matrix B:")
print(B)
print("Rank of B:", torch.linalg.matrix_rank(B))
```
